## 1. Smarter prices: if / elif / else (Lecture 4)

**Reminder**
* `elif` = "else, if...". Python checks the conditions **from top to bottom** and runs only the **first** one that is `True`.
* `else` catches everything that is left.
* Order matters: put the most specific condition first.

Coffee comes in three sizes:

````python
size = "M"

if size == "S":
    price = 10
elif size == "M":
    price = 12.5
elif size == "L":
    price = 15
else:
    price = 0
    print("Unknown size!")

print(f"A size {size} coffee costs {price:.2f} ILS")
````

**Order matters!** Pip gives a gift by order total. Which version is right?

````python
total = 120

# Version A: smallest condition first
if total >= 50:
    print("A: free cookie")
elif total >= 100:
    print("A: free cake")

# Version B: biggest condition first
if total >= 100:
    print("B: free cake")
elif total >= 50:
    print("B: free cookie")
````

In version A, `total >= 100` is never reached: 120 is already `>= 50`, so Python stops there. **Version B is right.**

**Decisions inside decisions:** on Fridays the café closes early, unless it is a holiday week.

````python
day = "Fri"
holiday_week = True

if day == "Fri":
    if holiday_week:
        print("Open until 20:00")
    else:
        print("Open until 14:00")
else:
    print("Open until 22:00")
````

## 2. Combining conditions: and / or / not (Lecture 4)

**Reminder**
* `and`: both must be `True`. `or`: at least one. `not`: flips `True` and `False`.
* Chaining: `8 <= hour < 10` means "hour is between 8 and 10".
* Text comparison is exact: `"Yes" == "yes"` is `False`.

Pip's discount rules:
* **Happy hour**: 16:00-18:00 → 20% off
* **Students in the morning** (before 10:00) → 15% off
* Everyone else, **unless** the café is full → a free refill

````python
hour = 9
student = True
cafe_full = False

if 16 <= hour < 18:
    print("Happy hour: 20% off")
elif student and hour < 10:
    print("Student morning: 15% off")
elif not cafe_full:
    print("Free refill today")
else:
    print("Regular price")
````

## 3. Under the hood: bits and logic gates (Lecture 4)

**Reminder**
* Computers store everything as **bits**: 0 and 1. `bin(13)` shows 13 in binary (`0b1101`).
* `int("1101", 2)` goes back from binary to a normal number.
* `and`, `or`, `not` are exactly the **logic gates** inside the chip.

The café door light: it turns on only if the door is open **and** it is dark (an AND gate).

````python
print(bin(13))
print(int("1101", 2))
print(bin(255))   # 8 bits, all of them 1

door_open = True
dark = False
print("Light on:", door_open and dark)   # AND gate
print("Alarm on:", door_open or dark)    # OR gate
print("Day mode:", not dark)             # NOT gate
````

## 4. The whole menu: lists (Lecture 5)

**Reminder**
* A list keeps many values, **in order**: `menu = ["Coffee", "Tea"]`.
* `append` adds at the end, `insert(i, x)` adds at position `i`, `remove(x)` removes by value, `pop()` takes the last one out.
* `x in my_list` asks "is it there?". `sort()` puts it in order.
* `len`, `sum`, `max`, `min` work on lists.

````python
menu = ["Coffee", "Tea", "Croissant"]
prices = [12.5, 8, 9]

menu.append("Cookie")
prices.append(4)

print(menu)
print(prices)
print("Items on the menu:", len(menu))
print("Cheapest:", min(prices), " Most expensive:", max(prices))
````

The **queue** at the counter is a list too. Savta Rina is let in first:

````python
queue = ["Noa", "Omer", "Dana"]
queue.append("Yossi")        # joins at the end
queue.insert(0, "Savta Rina") # goes first
print(queue)

served = queue.pop(0)         # the first person is served
print("Serving:", served)
print("Still waiting:", queue)

if "Dana" in queue:
    print("Dana is still waiting")
````

**Coffee sales this week** (Sunday to Thursday):

````python
cups = [42, 38, 51, 47, 60]

print("Total:", sum(cups))
print(f"Average per day: {sum(cups) / len(cups):.1f}")
print("Best day:", max(cups))
````

## 5. Regular customers: sets (Lecture 5)

**Reminder**
* A set keeps each value **once**, with no order: `{"Noa", "Omer"}`.
* `set(my_list)` drops the duplicates.
* `a & b` = in both, `a | b` = in at least one, `a - b` = in `a` but not in `b`.

Who came to the café on Sunday and on Monday?

````python
sunday_orders = ["Noa", "Omer", "Noa", "Dana", "Noa"]
print("Orders on Sunday:", len(sunday_orders))

sunday = set(sunday_orders)
print("Different customers on Sunday:", len(sunday))

monday = {"Dana", "Yossi", "Noa"}

print("Came both days:", sorted(sunday & monday))
print("Came at least once:", sorted(sunday | monday))
print("Only Sunday:", sorted(sunday - monday))
````

(`sorted(...)` only makes the printout tidy: a set has no order of its own.)

## 6. Positions: indexing and slicing (Lecture 5)

**Reminder**
* Positions start at **0**. `[-1]` is the last item.
* `[start:stop]` takes a slice. `stop` is **not** included.
* `[::-1]` reverses. Text can be sliced too.
* Lists can change, strings can't: `word[0] = "b"` is an error.

````python
orders = ["Latte", "Tea", "Espresso", "Latte", "Cookie", "Mocha", "Tea"]

print("First order:", orders[0])
print("Last order:", orders[-1])
print("Orders 2-4:", orders[1:4])
print("Last 3 orders:", orders[-3:])
print("Every second order:", orders[::2])
````

````python
item = "Croissant"
code_name = item[:3]
print(code_name)
print(item[::-1])
````

## 7. Repeat: while and for (Lecture 6)

**Reminder**
* `while condition:` repeats **as long as** the condition is `True`. Make sure something changes, or it never ends.
* `for x in something:` repeats once **for each** item of a list, a string or a `range`.
* `range(5)` → 0, 1, 2, 3, 4. `range(2, 10, 3)` → 2, 5, 8.
* `break` jumps out of the loop.

**while:** Noa has 50 ILS. How many coffees can she buy?

````python
money = 50
coffee_price = 12.5
coffees = 0

while money >= coffee_price:
    money = money - coffee_price
    coffees = coffees + 1

print(f"Coffees: {coffees}, money left: {money:.2f}")
````

**for:** print the whole menu. One loop replaces four copy-pasted `print` lines (Part 1 needed one line per item):

````python
for i in range(len(menu)):
    print(f"{menu[i]}: {prices[i]:.2f}")
````

**break:** search the queue and stop as soon as we find Dana:

````python
for person in queue:
    print("Checking", person)
    if person == "Dana":
        print("Found Dana!")
        break
````

## 8. The five loop recipes (Lecture 6)

Almost every loop is one of these five. Here they are, on one day of orders at Pip's.

````python
order_totals = [21.5, 9, 37.5, 12.5, 54, 8, 25]

# 1. Total: start at 0, add each one
total = 0
for amount in order_totals:
    total = total + amount

# 2. Average: total divided by how many
average = total / len(order_totals)

# 3. Count: start at 0, add 1 when the condition is True
big_orders = 0
for amount in order_totals:
    if amount >= 25:
        big_orders = big_orders + 1

# 4. Max: start with the first, replace when you find a bigger one
biggest = order_totals[0]
for amount in order_totals:
    if amount > biggest:
        biggest = amount

# 5. Search: a flag that starts False
found = False
for amount in order_totals:
    if amount > 50:
        found = True
        break

print(f"Total: {total:.2f}  Average: {average:.2f}")
print("Orders of 25 ILS or more:", big_orders)
print("Biggest order:", biggest)
print("Any order over 50?", found)
````

Python has shortcuts for some of them: `sum(order_totals)`, `max(order_totals)`, `50 in order_totals`.
Knowing the recipe means you can do things that have no shortcut, like **counting** the big orders.

## 9. All together: the counter program 2.0 🧾

Part 1's program knew 2 items and served 1 customer. Now it uses the whole menu (list), takes items **until the customer says `done`** (while),
looks up each price (for + search), and applies the discount rules (if / elif / and).

````python
# Pip's Café - counter program 2.0
menu = ["Coffee", "Tea", "Croissant", "Cookie"]
prices = [12.5, 8, 9, 4]

customer = input("Name: ")
student = input("Student? (yes/no) ") == "yes"
hour = int(input("Hour (0-23): "))

basket = []
item = input("Item (or done): ")
while item != "done":
    if item in menu:
        basket.append(item)
    else:
        print("  Sorry, we don't have", item)
    item = input("Item (or done): ")

# Look up each price with the search recipe
total = 0
print()
print(f"Receipt for {customer}")
for name in basket:
    for i in range(len(menu)):
        if menu[i] == name:
            print(f"  {name}: {prices[i]:.2f}")
            total = total + prices[i]

if 16 <= hour < 18:
    discount = 20
elif student and hour < 10:
    discount = 15
else:
    discount = 0

total = total * (100 - discount) / 100
print(f"  Discount: {discount}%")
print(f"  TOTAL:    {total:.2f} ILS")
print(f"  Items: {len(basket)}, different items: {len(set(basket))}")
````

## 10. Bug Zoo: every bug from Part 2 🐛

| Bug | What you see | Fix |
|---|---|---|
| `if x > 5` (no colon) | `SyntaxError: expected ':'` | add `:` |
| body of `if` / `for` / `while` not indented | `IndentationError` | indent the lines that belong to it |
| `if size = "M":` | `SyntaxError: invalid syntax` | `=` stores, `==` asks |
| `if input("Age? ") > 18:` | `TypeError: '>' not supported between 'str' and 'int'` | `int(input(...))` |
| `elif` conditions in the wrong order | no error, wrong answer! | most specific condition first |
| `queue[3]` on a list of 3 | `IndexError: list index out of range` | positions are 0, 1, 2 |
| `menu.remove("Pizza")` | `ValueError: list.remove(x): x not in list` | check with `in` first |
| `regulars[0]` on a set | `TypeError: 'set' object is not subscriptable` | sets have no positions. Use a list |
| `for i in 5:` | `TypeError: 'int' object is not iterable` | `for i in range(5):` |
| `total = 0` **inside** the loop | no error, wrong total! | start the total **before** the loop |
| `while` that never changes its condition | runs forever | change something inside. Stop it with ■ |

**Your turn:** this cell should print the total of the week. It has two bugs: one crashes, one gives a wrong answer.

````python
cups = [42, 38, 51, 47, 60]
for day in range(6):
    week_total = 0
    week_total = week_total + cups[day]
print(week_total)
````

2. `week_total = 0` is **inside** the loop, so it restarts every day. Move it above the loop. The right answer is `238`.
