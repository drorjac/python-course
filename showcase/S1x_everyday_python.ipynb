## 1. Your life in numbers

How many days have you lived? Hours? We ignore leap years to keep it simple (so the answer is a few days too small).

For the heartbeats we need an **assumption**: a resting heart beats roughly 60-100 times a minute. We take **70**.
The answer is only as good as the assumption - so we say it out loud.

```python
age = 20                 # try: age = int(input("Age? "))
days = age * 365         # no leap years
minutes = days * 24 * 60
beats_per_minute = 70    # ASSUMPTION
heartbeats = minutes * beats_per_minute
print(f"Days lived: about {days}")
print(f"Heartbeats: about {heartbeats:,}")
```

Big numbers are hard to read. Inside an f-string, `:,` adds a comma every three digits. Compare:

```python
hours = days * 24
print(f"Minutes without commas: {minutes}")
print(f"Minutes with commas: {minutes:,}")
print(f"Hours: {hours:,}")
```

### The billion-beat club

```python
beats_per_year = beats_per_minute * 60 * 24 * 365
billion_age = 1000000000 / beats_per_year       # age at 1 billion beats
print(f"You reach 1 billion heartbeats at about age {billion_age:.1f}")
```

```python
if heartbeats >= 1000000000:
    print("Welcome to the billion-beat club! 💓")
else:
    print(f"Not yet in the billion-beat club. {billion_age - age:.1f} years to go!")
```

**Try this:** put in your own age. Change `beats_per_minute` to 60, then 100. How much does the answer move?

## 2. A unit converter

1 mile is **exactly** 1.609344 km. A marathon is 42.195 km. How many miles is that?

```python
# try: km = float(input("km? "))
km_per_mile = 1.609344  # exact
km = 42.195             # a marathon
miles = km / km_per_mile
print(f"{km} km = {miles:.1f} miles")
```

And the other way round: miles to km means **multiply**.

```python
walk_miles = 3
walk_km = walk_miles * km_per_mile
print(f"{walk_miles} miles = {walk_km:.2f} km")
```

### Shekels to dollars

Exchange rates change **every day**. The rate below is only an **example** - check today's rate before you use it for real money!

```python
# try: shekels = float(input("NIS? "))
rate = 3.5     # EXAMPLE: NIS per dollar
shekels = 100
dollars = shekels / rate
print(f"{shekels} NIS = {dollars:.2f} USD")
```

```python
price_in_dollars = 25        # a T-shirt on a US website
price_in_shekels = price_in_dollars * rate
print(f"A {price_in_dollars}-dollar T-shirt = about {price_in_shekels:.2f} NIS (example rate)")
```

**Try this:** look up today's rate and put it in `rate`. Then convert your monthly coffee budget to dollars.

## 3. Splitting a restaurant bill

Four friends, one bill, a tip on top. Who pays what?

```python
bill = 340          # NIS (receipt)
num_friends = 4
tip_percent = 12    # change me!
tip = bill * tip_percent / 100
total = bill + tip
each = total / num_friends
print(f"Total: {total:.2f} NIS")
print(f"Each pays: {each:.2f} NIS")
```

```python
if each > 100:
    print("Pricey night! Falafel next time?")
else:
    print("Under 100 NIS each. Good deal!")
```

**Try this:** a fifth friend shows up at the end. Change `num_friends` to 5. Now try a 15% tip.

## 4. A little bit of luck: dice and a Magic 8-ball

Python comes with a ready-made tool for random numbers. `import random` means **"borrow a ready-made tool"** (more in Lecture 9).

`random.seed(7)` makes the "random" results the same every time you run the notebook, so we all see the same thing.
Delete that line for real surprises.

```python
import random       # borrow a tool
random.seed(7)      # same every run
roll = random.randint(1, 6)   # 1 to 6
print(f"You rolled a {roll}")
if roll == 6:
    print("Six! You win!")
else:
    print("Not a six. Roll again!")
```

Run the cell above again (without the seed line) a few times. Every number from 1 to 6 should show up sometimes.

### Magic 8-ball

Ask a yes/no question, get a (totally random) answer.
The square brackets hold a **list** of possible answers - lists are Lecture 5. `random.choice` picks one of them.

```python
question = "Will I pass the Python exam?"   # try: question = input("Ask me anything: ")
answers = ["Yes, definitely!", "Ask again later.", "Don't count on it.", "Signs point to yes."]

print(f"You asked: {question}")
print(f"The Magic 8-ball says: {random.choice(answers)}")
```
