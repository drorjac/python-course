## 1. A recipe with a name: def (Lecture 7)

**Reminder**
* You already used functions: `print`, `len`, `max`, `type`. Now you write your own.
* `def name(parameters):` defines the recipe. The indented lines are its body. **Nothing runs** until you **call** it: `name(...)`.
* Parameters are the blanks you fill in when you call it. Order matters.

````python
def welcome(name):
    print(f"Hi {name}, welcome to Pip's Café! ☕")

welcome("Noa")
welcome("Omer")
````

````python
def order_coffee(size, milk):
    print(f"One {size} coffee with {milk} milk")

order_coffee("large", "oat")
order_coffee("oat", "large")   # order matters!
````

A function can hold a loop. Print the menu with one call, wherever we need it:

````python
def print_menu(items, prices):
    print("--- Pip's Café ---")
    for i in range(len(items)):
        print(f"{items[i]}: {prices[i]:.2f}")

print_menu(["Coffee", "Tea", "Croissant"], [12.5, 8, 9])
````

## 2. Getting an answer back: return (Lecture 7)

**Reminder**
* `return value` sends an answer back to whoever called the function. You can store it, print it, or use it in math.
* `print` only **shows** a value. A function without `return` gives back `None`.
* `return` ends the function right away.
* Default values: `def f(x, tip=10):` means "if you don't say, tip is 10".

````python
def with_tip(bill, tip_percent=10):
    return bill + bill * tip_percent / 100

dinner = with_tip(80)          # default 10%
lunch = with_tip(50, 15)       # 15%
print(dinner, lunch)
print(f"Both together: {dinner + lunch:.2f}")
````

**return vs print.** Both functions *look* the same when you call them, but only one gives back an answer:

````python
def double_print(x):
    print(x * 2)

def double_return(x):
    return x * 2

a = double_print(6)
b = double_return(6)
print("a is", a)
print("b is", b)
````

`a` is `None`: `double_print` only showed 12 on the screen. Use `return` whenever the answer is needed later.

**The bricks put together:** small functions that call each other.

````python
def average(numbers):
    return sum(numbers) / len(numbers)

def week_report(name, cups):
    print(f"{name}: average {average(cups):.1f} cups a day, best day {max(cups)}")

week_report("Coffee", [42, 38, 51, 47, 60])
week_report("Tea", [20, 25, 18, 30, 22])
````

## 3. What stays inside: scope (Lecture 7)

**Reminder**
* A variable created **inside** a function is **local**: it lives only while the function runs.
* Inside can **read** a variable from outside (global).
* The same name inside and outside = two different boxes.
* To get a value out, use `return`.

````python
vat = 0.18   # global: the whole notebook sees it

def price_with_vat(price):
    total = price + price * vat   # reads the global vat
    return total

print(price_with_vat(100))
print(total)   # total was local: it is gone
````

`NameError: name 'total' is not defined`: `total` lived only inside the function. The fix: keep the returned value, `result = price_with_vat(100)`.

## 4. The menu as a dictionary (Lecture 8) 🔁

**Reminder**
* A dictionary maps **keys** to **values**: `menu = {"Coffee": 12.5, "Tea": 8}`.
* Look up: `menu["Tea"]`. Add or change: `menu["Cookie"] = 4`.
* `"Tea" in menu` checks for a key. `menu.get("Pizza", 0)` gives a default instead of an error.
* Loop over pairs: `for item, price in menu.items():`

**Refactor moment.** In Part 2 the menu was two lists that had to stay in the same order:
```python
menu = ["Coffee", "Tea", "Croissant", "Cookie"]
prices = [12.5, 8, 9, 4]
```
One dictionary does the same job, and the price can never get out of step with its item:

````python
menu = {"Coffee": 12.5, "Tea": 8, "Croissant": 9, "Cookie": 4}

print(menu["Croissant"])
menu["Muffin"] = 11          # add
menu["Tea"] = 8.5            # change
print(menu)

print("Pizza" in menu)
print(menu.get("Pizza", 0))
````

````python
for item, price in menu.items():
    print(f"{item}: {price:.2f}")

print("Everything once:", sum(menu.values()))
````

**Counting with a dictionary.** Which item sold the most today?

````python
sold = ["Coffee", "Tea", "Coffee", "Muffin", "Coffee", "Tea", "Croissant"]

counts = {}
for item in sold:
    counts[item] = counts.get(item, 0) + 1

print(counts)
````

`counts.get(item, 0)` means "how many so far, or 0 if this is the first one".

## 5. Tables: lists inside lists, dicts inside lists (Lecture 8)

**Reminder**
* A table is a **list of rows**. `tables[row][seat]`: first the row, then the item inside it.
* A loop inside a loop visits every cell.
* A **list of dictionaries** is a table with named columns: each dict is one row.

The café has 3 tables with 4 seats. `""` is a free seat:

````python
tables = [
    ["Noa", "Omer", "", ""],
    ["Dana", "", "", "Yossi"],
    ["", "", "", ""],
]

print("Second table, last seat:", tables[1][3])   # positions start at 0

free = 0
for row in tables:
    for seat in row:
        if seat == "":
            free = free + 1
print("Free seats:", free)
````

## 6. The average can lie: median and quantiles (Lecture 8)

**Reminder**
* The **median** is the middle value after sorting. One huge value can pull the average far away, but not the median.
* A **quantile** `q` is the value that a share `q` of the data is below. The median is the 0.5 quantile.
* `sorted(values)` gives a sorted **copy**.

The same functions we wrote in Lecture 8:

````python
def median(values):
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:     # odd count
        return ordered[middle]
    else:                         # even count
        return (ordered[middle - 1] + ordered[middle]) / 2

def quantile(values, q):
    ordered = sorted(values)
    position = int(q * (len(ordered) - 1))
    return ordered[position]
````

On Friday, a company ordered coffee for the whole office: one order of 450 ILS. What does a **typical** order look like?

````python
friday = [21.5, 9, 37.5, 12.5, 450, 8, 25]

print(f"Average: {sum(friday) / len(friday):.2f}")
print("Median: ", median(friday))
print("0.25 quantile:", quantile(friday, 0.25))
print("0.75 quantile:", quantile(friday, 0.75))
````

The average (80+) describes nobody. The median (21.5) is a typical order. That's why house prices and salaries are reported as medians.

## 7. Debugging like a pro (Lecture 8)

**Reminder**
* Read the error **from the bottom up**: the last line says *what* went wrong, the lines above say *where*.
* The common ones: `NameError`, `TypeError`, `ValueError`, `IndexError`, `KeyError`, `ZeroDivisionError`.
* A **logic bug** gives no error, just a wrong answer. Print the values inside the loop to see where it goes wrong.

On a quiet day there are no orders yet. Read the error from the bottom up:

````python
quiet_day = []
print(average(quiet_day))
````

Bottom line: `ZeroDivisionError`. One line up: inside `average`, at `sum(numbers) / len(numbers)`. `len([])` is 0.
Fix: check first, e.g. `if len(numbers) == 0: return 0`.

**A logic bug.** Pip's loyalty rule: every 5th coffee is free. This function gives the wrong answer for 10 coffees (should be 2 free):

````python
def free_coffees(count):
    free = 0
    for n in range(1, count):
        if n % 5 == 0:
            free = free + 1
    return free

print(free_coffees(10))
````

**Print debugging:** show what the loop really does.

````python
def free_coffees(count):
    free = 0
    for n in range(1, count):
        print("coffee number", n)   # debug line
        if n % 5 == 0:
            free = free + 1
    return free

print("Free:", free_coffees(10))
````

Coffee number 10 never appears: `range(1, count)` stops **before** `count`. The fix is `range(1, count + 1)`:

````python
def free_coffees(count):
    free = 0
    for n in range(1, count + 1):
        if n % 5 == 0:
            free = free + 1
    return free

print(free_coffees(10), free_coffees(4), free_coffees(5))
````

## 8. Coding with AI: always test (Lecture 8)

**Reminder**
* AI can be **confidently wrong**. The code looks nice and runs, but the answer is wrong.
* Test with cases where you **know** the answer, including edge cases (empty, one item, a tie, a boundary like exactly 60).
* A good prompt: who you are, what you want, the input and output, and an example.

We asked an AI for a median function. It wrote this:

````python
def ai_median(values):
    middle = len(values) // 2
    return values[middle]

# Tests: (input, expected answer)
print(ai_median([1, 2, 3]), "expected 2")
print(ai_median([3, 1, 2]), "expected 2")
print(ai_median([1, 2, 3, 4]), "expected 2.5")
````

The first test passes, which is exactly why one test is not enough. The AI forgot to **sort**, and forgot the **even** case.
Our own `median` passes all three:

````python
print(median([1, 2, 3]), median([3, 1, 2]), median([1, 2, 3, 4]))
````

## 9. All together: the counter program 3.0 🧾

Part 2's program was one long block with two lists. Now each job is a small function, the menu is a dictionary,
and the same functions produce the whole daily report.

````python
# Pip's Café - counter program 3.0
menu = {"Coffee": 12.5, "Tea": 8, "Croissant": 9, "Cookie": 4, "Muffin": 11}


def order_total(items, discount=0):
    total = 0
    for item in items:
        total = total + menu.get(item, 0)
    return total * (100 - discount) / 100


def print_receipt(customer, items, discount=0):
    print(f"Receipt for {customer}")
    for item in items:
        print(f"  {item}: {menu[item]:.2f}")
    if discount > 0:
        print(f"  Discount: {discount}%")
    print(f"  TOTAL: {order_total(items, discount):.2f} ILS")


def daily_report(orders):
    totals = []
    counts = {}
    for order in orders:
        totals.append(order_total(order["items"]))
        for item in order["items"]:
            counts[item] = counts.get(item, 0) + 1

    best_item = ""
    for item, n in counts.items():
        if best_item == "" or n > counts[best_item]:
            best_item = item

    print(f"Orders: {len(orders)}")
    print(f"Income: {sum(totals):.2f} ILS")
    print(f"Median order: {median(totals):.2f} ILS")
    print(f"Best seller: {best_item} ({counts[best_item]} sold)")


print_receipt("Dana", ["Coffee", "Coffee", "Muffin"], discount=15)
print()
daily_report(orders)
````

## 10. Bug Zoo: every bug from Part 3 🐛

| Bug | What you see | Fix |
|---|---|---|
| calling a function **above** its `def` | `NameError` | define first, then call |
| `welcome()` without the argument | `TypeError: missing 1 required positional argument` | pass every parameter |
| forgot `return`, then `result + 20` | `TypeError: unsupported operand ... 'NoneType'` | `return` the answer, don't `print` it |
| using a local variable outside | `NameError` | `return` it and store the result |
| `menu["Falafel"]` | `KeyError: 'Falafel'` | check with `in`, or `menu.get("Falafel", 0)` |
| `for item, price in menu:` | `ValueError: too many values to unpack` | `menu.items()` |
| `tables[3][0]` on 3 rows | `IndexError` | rows are 0, 1, 2 |
| `average([])` | `ZeroDivisionError` | check for an empty list first |
| `range(1, count)` misses the last one | no error, wrong answer! | `range(1, count + 1)` |
| AI code that passed one test | no error, wrong answer! | test several cases, including edge cases |

**Your turn:** this function should return the price of a basket with 18% VAT. Two bugs. Run, read the error, fix, repeat.
Check: `basket_price(["Tea", "Cookie"])` should be `14.16`.

````python
def basket_price(items):
    total = 0
    for item in items:
        total = total + menu[item]
    print(total * 1.18)

result = basket_price(["Tea", "cookie"])
print(f"{result:.2f}")
````

<details><summary>Show the fixes</summary>

1. `KeyError: 'cookie'`: keys are exact, the menu has `"Cookie"` with a capital C.
2. Then `TypeError ... NoneType`: the function **prints** instead of **returning**. Change `print(total * 1.18)` to `return total * 1.18`.
