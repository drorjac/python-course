## 1. The sign on the door (Lecture 1)

**Reminder**
* `print(...)` shows something on the screen.
* Text goes inside quotes `"..."`. Without quotes, Python **calculates**.
* `#` starts a **comment**: a note for humans. Python skips it.

```python
# The sign on Pip's door
print("Welcome to Pip's Café!")
print("Open every day")
print("Hours open per week:")
print(7 * 10)   # 7 days, 10 hours a day: Python does the math
```

**Text or math?** Quotes change everything. The first line prints the *text* `7 * 10`, the second prints the *answer*.

```python
print("7 * 10")
print(7 * 10)
```

## 2. The menu: variables (Lecture 2)

**Reminder**
* A variable is a **labeled box**: `price = 12` puts 12 in a box called `price`.
* `=` means "put into the box", not "is equal".
* A box holds **one value at a time**. A new value replaces the old one.
* Good names tell a story: `coffee_price`, not `x`.

```python
coffee_price = 12.5
croissant_price = 9
cookie_price = 4

print("Coffee:", coffee_price)
print("Croissant:", croissant_price)
print("Cookie:", cookie_price)
```

Variables can **work together**. Noa orders a coffee and a croissant:

```python
noa_total = coffee_price + croissant_price
print(noa_total)
```

**Update a box using its own value.** Pip baked 30 cookies. Every sale takes one away:

```python
cookies_left = 30
cookies_left = cookies_left - 1   # one sold
cookies_left = cookies_left - 1   # another one
print(cookies_left)
```

💡 **Try it:** the price of coffee goes up by 1.5. Update `coffee_price` using its own value and print it.

## 3. What kind of value? Types (Lecture 2)

**Reminder**

| Type | Example | In the café |
|---|---|---|

| `float` | `12.5` | a price with cents |
| `str` | `"Latte"` | the name of a drink |

`type(...)` asks Python what kind of value something is.

**Quotes change everything:** `+` adds numbers, but **glues** text.

```python
print(5 + 5)
print("5" + "5")
print(len(drink))   # how many letters in "Latte"
```

## 4. Café math (Lecture 2)

**Reminder**

| Symbol | Meaning | Example | Result |
|---|---|---|---|
| `+ - *` | plus, minus, times | `3 * 4` | `12` |
| `/` | divide (always a `float`) | `10 / 2` | `5.0` |
| `//` | divide, keep only whole part | `50 // 6` | `8` |
| `%` | what is left over | `50 % 6` | `2` |
| `**` | power | `2 ** 3` | `8` |

Pip baked 50 cookies and packs them in boxes of 6. How many full boxes, and how many cookies are left?

```python
cookies = 50
box_size = 6

full_boxes = cookies // box_size
leftover = cookies % box_size

print("Full boxes:", full_boxes)
print("Left over:", leftover)
```

Pip's coffee machine was on for 610 minutes today. How many hours and minutes is that?

```python
machine_minutes = 610
print(machine_minutes // 60, "hours and", machine_minutes % 60, "minutes")
```

**Order of operations** matters. Three friends share a bill of one coffee each plus one croissant to share:

```python
wrong = 3 * coffee_price + croissant_price / 3
right = (3 * coffee_price + croissant_price) / 3

print("Without parentheses:", wrong)
print("With parentheses:   ", right)
```

Only the second line is "the bill split by 3". Python does `*` and `/` before `+`, just like in math class.
Strings can do `*` too:

```python
print("☕" * 5)
print("-" * 20)
```

## 5. Talking to the customer: print and input (Lecture 3)

**Reminder**
* `print` can show several things, separated by commas. `sep=` chooses the glue between them, `end=` what comes last.
* `input("question")` waits for the user to type, and gives back what they typed.
* ⚠️ `input` **always gives text** (`str`), even when the user types a number.

```python
print("Coffee", "Croissant", "Cookie", sep=" | ")
print("Opening day", end="! ")
print("Everything 10% off")
```

```python
name = input("What is your name? ")
print("Hi", name, "- welcome to Pip's Café!")
```

```python
answer = input("How many cookies? ")
print(type(answer))
print(answer + answer)   # text glue, not math!
```

Typing `2` gave the **text** `"2"`, so `+` glued it: `22`. To do math we need to **convert** first.

## 6. The receipt: converting types and f-strings (Lecture 3)

**Reminder**
* `int("2")` → `2`, `float("12.5")` → `12.5`, `str(9)` → `"9"`.
* `int(3.9)` → `3`: `int` **chops**, it does not round.
* Convert the answer right away: `count = int(input("How many? "))`.
* An **f-string** puts values inside text: `f"Total: {total}"`.
* `{price:.2f}` shows exactly 2 digits after the point. Money always needs it.

```python
print(int("2") + int("2"))
print(int(3.9))
print(float("12.5") * 2)
```

```python
item = "Coffee"
count = 3
total = count * coffee_price

print(f"{count} x {item} = {total} NIS")
print(f"{count} x {item} = {total:.2f} NIS")
```

**Tip calculator.** The bill is 47.5 NIS and the customer leaves a 12% tip:

```python
bill = 47.5
tip = bill * 12 / 100
print(f"Bill: {bill:.2f}")
print(f"Tip:  {tip:.2f}")
print(f"Pay:  {bill + tip:.2f}")
```

## 7. Decisions: a first if / else (Lecture 3)

**Reminder**
* Comparisons give `True` or `False`: `==  !=  <  >  <=  >=`.
* `=` puts a value in a box. `==` **asks** "are these equal?".
* `if` needs a colon `:`, and the lines that belong to it are **indented**.

Pip's rule: an order of 50 NIS or more gets a free cookie.

```python
order_total = 53

print(order_total >= 50)

if order_total >= 50:
    print("You get a free cookie! 🍪")
else:
    print("Spend", 50 - order_total, "more for a free cookie")
```

**Even or odd?** Tables with an even number are by the window. `%` tells us:

```python
table = 7
if table % 2 == 0:
    print("Window table")
else:
    print("Garden table")
```

💡 **Try it:** change `order_total` to `49`, then `50`. Change `table` to `12`. Predict first!

## 8. All together: the counter program 🧾

Everything from Part 1 in one small program: `print`, comments, variables, types, math, `input`, `int()`, f-strings and `if` / `else`.
Run it and answer the questions.

```python
# Pip's Café - the counter program
coffee_price = 12.5
croissant_price = 9
student_discount = 10   # percent

customer = input("Name: ")
coffees = int(input("How many coffees? "))
croissants = int(input("How many croissants? "))
student = input("Student? (yes/no) ")

total = coffees * coffee_price + croissants * croissant_price

print()
print("=" * 28)
print("       Pip's Café ☕")
print("=" * 28)
print(f"Customer: {customer}")
print(f"{coffees} x Coffee     {coffees * coffee_price:8.2f}")
print(f"{croissants} x Croissant  {croissants * croissant_price:8.2f}")

if student == "yes":
    discount = total * student_discount / 100
    total = total - discount
    print(f"Student -{student_discount}%  {-discount:9.2f}")
else:
    print("No discount")

print("-" * 28)
print(f"TOTAL          {total:8.2f} NIS")
print("=" * 28)
```

`{x:8.2f}` = 2 digits after the point, and at least 8 characters wide, so the numbers line up in a column.

## 9. Bug Zoo: every bug from Part 1 🐛

Every programmer sees these. The trick is to **read the last line** of the error message.

| Bug | Error you see | Fix |
|---|---|---|
| `Print("hi")` | `NameError: name 'Print' is not defined` | Python is case sensitive: `print` |
| `print("hi)` | `SyntaxError: unterminated string literal` | close the quote |
| `print(totl)` | `NameError: name 'totl' is not defined` | spell the variable exactly as it was created |
| `2nd_place = "Dana"` | `SyntaxError: invalid decimal literal` | names cannot start with a digit: `second_place` |
| `my score = 90` | `SyntaxError: invalid syntax` | no spaces in names: `my_score` |
| `"Age: " + 20` | `TypeError: can only concatenate str (not "int") to str` | `"Age: " + str(20)`, or an f-string |
| `input("Count? ") + 1` | `TypeError` | convert first: `int(input("Count? ")) + 1` |
| `int("3.5")` | `ValueError: invalid literal for int()` | `float("3.5")` |
| `if total > 50` | `SyntaxError: expected ':'` | add the colon |
| `if` body not indented | `IndentationError: expected an indented block` | indent the lines that belong to the `if` |
