# L01-03: Comments, notes for humans
Lecture 1, slides: *Comments: notes for humans*

Code is read by two readers: the computer and **people** (including you, next week, at 2 a.m.).
A comment is a note for the people. The intern skips it completely.
In Python, everything after a `#` on a line is a comment.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## A comment on its own line, and at the end of a line
🔮 **Predict first:** how many lines of output will you see?

```python
# This line is a comment - Python ignores it
print("Hi!")   # a comment can also go here
```

Only `Hi!` appears. Use comments to explain **why** you wrote something, for future you.

## Commenting a line out
A handy trick: put a `#` in front of a line of code to switch it **off** without deleting it.
Remove the `#` later to switch it back on.

🔮 **Predict first:** which lines will be printed?

```python
print("Line one")
# print("Line two")
print("Line three")
```

`Line two` is not printed. The intern sees the `#` and skips the whole line.
