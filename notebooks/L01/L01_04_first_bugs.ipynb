## Bug 1: capital letters matter
🔮 **Predict first:** will this work?

```python
Print("hi")
```

**What went wrong:** Python sees `Print` and `print` as two different words.
It only knows `print`, all small letters.
The last line says `NameError: name 'Print' is not defined`.
(Plain Python, like on the slide, even adds `Did you mean: 'print'?`. Notebooks sometimes leave that hint out.)

**Fix:** write `print` with a small `p`.

```python
print("hi")
```

## Bug 2: a missing quote
🔮 **Predict first:** spot the problem before you run it.

```python
print("hi)
```

**What went wrong:** the text starts with `"` but never ends.
Python reads to the end of the line, looking for the closing quote, and gives up:
`SyntaxError: unterminated string literal` ("unterminated" means "never closed").

**Fix:** close the quote.

```python
print("hi")
```

## Bug 3: missing parentheses
🔮 **Predict first:** what is missing here?

```python
print hi
```

**What went wrong:** `print` needs parentheses `( )` around what it should show.
And text needs quotes. Python says: `Missing parentheses in call to 'print'. Did you mean print(...)?`

**Fix:** add the parentheses and the quotes.

```python
print("hi")
```
