# L01-02: print shows text and math
Lecture 1, slides: *print can show text and math*, *Predict the output*

`print` can show words, and it can also do math for you.
The tiny detail that decides which one you get: **quotes**.
Let's see how the literal intern reads them.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## Text or math?
🔮 **Predict first:** what will each line show? Write your guess on paper, then run.

```python
print("I love pizza")
print(2 + 3)
print("2 + 3")
```

**Notice:** no quotes, Python calculates. With quotes, Python just copies the text.

## Predict the output
`*` means "times". 🔮 **Predict first**, then run.

```python
print(3 * 4)
print("3 * 4")
```

Line 1 has no quotes, so the intern multiplies: `12`.
Line 2 has quotes, so the intern copies the text exactly: `3 * 4`.
