## The math symbols

| Symbol | Meaning | Example |
|---|---|---|
| `+ -` | plus, minus | `7 - 3` gives `4` |
| `*` | times | `7 * 3` gives `21` |
| `/` | divide | `7 / 2` gives `3.5` |
| `//` | divide, drop the rest | `7 // 2` gives `3` |
| `%` | the leftover (remainder) | `7 % 2` gives `1` |
| `**` | power | `7 ** 2` gives `49` |

The examples from the table, in code:

```python
print(7 - 3)
print(7 * 3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
```

## `/` always gives a float
🔮 **Predict first:** `10 / 2` divides evenly. Will you see `5` or `5.0`?

```python
print(10 / 2)
print(9 / 4)
```

Even when it divides evenly you get `5.0`, not `5`. Python plays it safe: the answer might need decimals.

## Split the pizza: `//` and `%`
8 slices, 3 friends. `//` tells you how many slices each friend gets. `%` tells you how many are left over.

```python
slices = 8
friends = 3
print(slices // friends)
print(slices % friends)
```

## How long is the movie?
The app says 135 minutes. How many hours and minutes is that?

```python
movie_minutes = 135
print(movie_minutes // 60)  # hours
print(movie_minutes % 60)   # minutes
```

135 minutes = **2 h 15 min**. 🍿

## Who goes first?
Same order as in school math:
1. Parentheses `( )`
2. Power `**`
3. `*` `/` `//` `%`
4. `+` `-`

🔮 **Predict first:** are the two lines the same?

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
```

Not sure who goes first? **Add parentheses.**

## Average of three exam grades
Grades 85, 92 and 78. 🔮 **Predict first:** one of these lines is wrong. Which one?

```python
print(85 + 92 + 78 / 3)
print((85 + 92 + 78) / 3)
```

Line 1 divides **only the 78**. Oops! An average of 203 would be a nice grade, but it's not real.
Line 2 adds first (parentheses), then divides: `85.0`.

## Café bill for three friends

```python
coffee = 14
croissant = 12.5
num_friends = 3
bill = num_friends * (coffee + croissant)
print(bill)
```

Now change `num_friends` to 5 and run the cell again. No retyping of the math!

## Power: `**`
`2 ** 10` means "double 1, ten times". `side ** 2` is the area of a 4 by 4 square.

```python
print(2 ** 10)
side = 4
print(side ** 2)
```

## Strings can do `+` and `*` too
`+` glues strings together. `*` repeats a string.
🔮 **Predict first:** what does `"ha" * 3` give?

```python
first = "Tel"
second = "Aviv"
print(first + " " + second)
print("ha" * 3)
print("-" * 20)
```
