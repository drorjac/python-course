### Solution 2: The pizza party planner

69 slices need 9 pizzas (8.6 rounds up), costing 468 NIS, about 20.35 NIS each.

```python
students = 23
slices_each = 3
slices_needed = students * slices_each
pizzas = (slices_needed + 7) // 8
cost = pizzas * 52
print(slices_needed)
print(pizzas)
print(cost)
print(cost / students)  # per student
```
