Values come in different kinds: whole numbers, decimal numbers, text, and True/False.
The **type** decides what you can do with a value. You can add prices, but you can't add "Dana" and 3.
Let's meet the four types and ask Python which is which.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## Numbers: `int` and `float`
`int` is a whole number (also negative), like students in a class.
`float` has a decimal point, like the price of a coffee.

```python
num_students = 30     # int
temperature = -3      # int
price = 12.5          # float
print(price)
```

## Text: `str` (string)
A string is text in quotes, `"double"` or `'single'`.
Does the text have an apostrophe, like *Dana's*? Use double quotes.

🔮 **Predict first:** do the quotes appear in the output?

```python
city = "Tel Aviv"
greeting = 'Shalom!'
quote = "Dana's phone"
print(city)
print(greeting)
print(quote)
```

## True or False: `bool`
Like a light switch: only **on** or **off**. Is it raining? Do I have an exam?
Capital `T` and `F`, no quotes! (Named after George Boole, a 19th-century mathematician.)

```python
is_raining = True
has_exam = False
print(is_raining)
```

## Ask Python: `type()`
`type()` tells you the kind of any value. Read `class` as "kind".

```python
print(type(30))
print(type(12.5))
print(type("Dana"))
print(type(True))
```

## Quotes change everything
🔮 **Predict first:** what does `"5" + "5"` give?

```python
print(5 + 5)
print("5" + "5")
print(type("5"))
```

`"5"` is not a number. It is the **character** 5, like the number printed on a T-shirt.
Two pieces of text with `+` are glued together: `55`.

## One letter is just a short string
Python has no special "letter" type. A letter is a string of length 1.
`len()` counts the characters.

```python
letter = "A"
print(type(letter))
print(len(letter))
print(len("pizza"))
```
