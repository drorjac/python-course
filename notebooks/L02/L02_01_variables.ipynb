## Your first variable
Line 1 puts `12` into a box called `price`. Line 2 has no quotes, so Python **opens the box** and shows what is inside.

Read `=` as "**gets**": "price gets 12". It means "store", not "equals" like in math class.

```python
price = 12
print(price)
```

## Variables can work together
A coffee and a croissant. Python opens both boxes, adds the values, and stores the result in a new box.

```python
coffee = 14
croissant = 12
total = coffee + croissant
print(total)
```

## A box holds one value at a time
Someone liked your post! When you store a new value, the old one is thrown away.

🔮 **Predict first:** what are the two printed numbers?

```python
num_likes = 10
print(num_likes)
num_likes = 11
print(num_likes)
```

## Update a box using its own value
Like a game score. **Right side first:** `0 + 10` is `10`, then it is stored back in `score`.

🔮 **Predict first:** what is the final score?

```python
score = 0
score = score + 10
score = score + 5
print(score)
```

## Predict the output
🔮 **Predict first:** `b = a` copies... what exactly? Write your guess, then run.

```python
a = 5
b = a
a = 10
print(b)
```

`5`! `b = a` copies the value **at that moment**. Changing `a` later does not touch `b`.
