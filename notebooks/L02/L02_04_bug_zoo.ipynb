## Bug 1: mixing text and numbers
🔮 **Predict first:** will Python glue `"Age: "` and `20` together?

```python
age = 20
print("Age: " + age)
```

**What went wrong:** "Concatenate" means "glue". `"Age: "` is text (`str`), `age` is a number (`int`).
You can glue text only to text.

**Fix for now:** use text on both sides. (Next lecture you will learn a better fix.)

```python
print("Age: " + "20")
```

## Bug 2: a typo in the name
🔮 **Predict first:** look carefully at the second line.

```python
total = 26
print(totl)
```

**What went wrong:** there is no box called `totl`. Python only knows `total`.
Plain Python, like on the slide, even suggests the fix: `Did you mean: 'total'?` (Notebooks sometimes leave that hint out.)

**Fix:** correct the typo.

```python
total = 26
print(total)
```

## Bug 3: capitals matter

```python
price = 12
print(Price)
```

**What went wrong:** `price` and `Price` are two different boxes. Only `price` exists.

**Fix:** use exactly the same name, small `p`.

```python
price = 12
print(price)
```

## Bug 4: a name that starts with a digit

```python
2nd_place = "Dana"
```

**What went wrong:** names cannot start with a digit. Python saw a `2` and expected a number,
so it complains about an "invalid decimal literal" (a broken number).

**Fix:** start with a letter.

```python
second_place = "Dana"
print(second_place)
```

## Bug 5: a space inside a name

```python
my score = 90
```

**What went wrong:** a name cannot contain spaces. Python reads `my` and `score` as two separate words and gets lost.

**Fix:** use an underscore `_` instead of the space.

```python
my_score = 90
print(my_score)
```
