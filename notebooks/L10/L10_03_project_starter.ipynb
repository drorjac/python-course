## Step 3b: a small class of your own (Lecture 9)
A class keeps one group's data and its actions together. This one holds the values of one group
and can report on itself. Change the name and the methods to fit your data.

```python
class Group:
    def __init__(self, name, values):
        self.name = name
        self.values = values

    def average(self):
        return round(sum(self.values) / len(self.values), 2)

    def report(self):
        print(f"{self.name}: {len(self.values)} rows, average {self.average()}")

chosen = Group(one_value, list(group[number_column]))  # 👉 YOUR PROJECT HERE
everyone = Group("Everyone", list(df[number_column]))
chosen.report()
everyone.report()
```
