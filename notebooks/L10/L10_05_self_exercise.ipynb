### Solution 1: Plot the menu


```python
import matplotlib.pyplot as plt

menu = {"coffee": 12, "tea": 10, "cake": 18}
plt.bar(menu.keys(), menu.values())
plt.title('Menu prices')
plt.ylabel('ILS')
plt.show()
```

### Solution 2: A tiny regression

The pattern is `y = 2x`, so the model should predict close to 10.

```python
from sklearn.linear_model import LinearRegression

x = [[1], [2], [3], [4]]
y = [2, 4, 6, 8]
model = LinearRegression()
model.fit(x, y)
print(model.predict([[5]]))
```

### Solution 3: Receipt as a class

Same idea as the Lecture 8 menu total, now packaged as an object with its own data and behavior.

```python
class Receipt:
    def __init__(self, items):
        self.items = items

    def total(self):
        return sum(self.items.values())

r = Receipt({'coffee': 12, 'cake': 18})
print(r.total())
```
