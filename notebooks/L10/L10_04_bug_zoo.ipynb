## Bug 1: x and y of different length
🔮 **Predict first:** count the items in each list.

## Bug 2: a typo in a command name

## Bug 3: the 2-D surprise

```python
from sklearn.linear_model import LinearRegression

hours = [1, 2, 3]
grades = [60, 70, 80]
model = LinearRegression()
model.fit(hours, grades)
```

**What went wrong:** `ValueError: Expected 2D array, got 1D array instead:` (the first line of a longer error message). For scikit-learn, the examples are always **a table**, even with one column.

**Fix:** one inner list per example: `[[1], [2], [3]]`.

```python
hours = [[1], [2], [3]]
grades = [60, 70, 80]
model = LinearRegression()
model.fit(hours, grades)
print(model.predict([[4]]))
```
