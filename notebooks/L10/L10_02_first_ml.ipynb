Until now **you** wrote the rules. In machine learning the computer gets examples (and sometimes the right answers) and **learns** the rule by itself. 🤖 We use **scikit-learn** (`sklearn`), free and already installed in Colab. Every model follows the same 3 steps:

1. `model = SomeModel()` (create)

3. `model.predict(new_examples)` (answer new exam questions)

## 1. Regression: predict a grade
Predict a **number**. Examples: hours studied. Answers: exam grades. Note the **double brackets** `[[1], [2], ...]`: the examples are a table with one column.

```python
from sklearn.linear_model import LinearRegression

hours = [[1], [2], [3], [4], [5], [6], [8]]
grades = [52, 58, 60, 68, 72, 78, 88]
model = LinearRegression()
model.fit(hours, grades)
print(model.predict([[7]]))
```

7 hours of study: about 83.

## 2. What did the model learn? A line!
Same notebook, next cell: we reuse `hours`, `grades` and `model` from above. The "rule" it learned is the best straight line through the dots.

```python
import matplotlib.pyplot as plt

plt.scatter(hours, grades)
plt.plot(hours, model.predict(hours),
         color="coral")
plt.xlabel("Hours studied")
plt.ylabel("Exam grade")
plt.show()
```

## 3. Careful: models don't know common sense
🔮 **Predict first:** what grade does the model give to someone who studied 30 hours?

```python
print(model.predict([[30]]))
```

A grade of 201?! ⚠️ The model only saw 1 to 8 hours. Far outside its examples, it just continues the line. Always ask: does this answer make sense?

## 4. Classification: apple or banana?
Pick a **label**. Each fruit has a **weight** (grams) and a **redness** (0 = yellow, 10 = red). We give 8 examples with the right label. For a new fruit, the model looks at its **3 nearest** examples, and the majority vote wins. (Made-up fruit.)

```python
from sklearn.neighbors import KNeighborsClassifier

fruits = [[120, 2], [130, 1], [115, 3], [125, 2],
          [170, 8], [160, 9], [180, 7], [150, 8]]
labels = ["banana", "banana", "banana", "banana",
          "apple", "apple", "apple", "apple"]
model = KNeighborsClassifier(n_neighbors=3)
model.fit(fruits, labels)
print(model.predict([[155, 7], [118, 3]]))
```

Same 3 steps: create, `fit`, `predict`. (Careful: `model` is now the fruit model, not the grade model.)

## 5. Clustering: find groups by itself
Find **groups**, with **no answers given**. 30 café customers: how many times they visit per month, and how much they spend on average (ILS). `random_state=0` means: the same result every run.

📁 **Before you start:** upload `notebooks/L10/data/cafe_customers.csv` to Colab first (click the **folder icon** on the left, then the upload button). No file? No panic: the Plan B cell below builds the same table for you.

```python
import os
import pandas as pd

# Plan B: only needed if you could NOT upload the file.
if os.path.exists("cafe_customers.csv"):
    print("Found cafe_customers.csv. Great, you are ready!")
else:
    print("No file found. Plan B: building the same table from a dictionary.")
    backup = {
        "visits": [20, 22, 16, 7, 7, 8, 1, 9, 7, 8, 3, 10, 21, 1, 16,
                   16, 3, 3, 10, 3, 11, 8, 21, 2, 22, 1, 5, 1, 21, 16],
        "spend": [16, 10, 10, 27, 20, 23, 45, 27, 30, 30, 36, 29, 13, 47, 10,
                  9, 41, 42, 25, 42, 20, 28, 12, 44, 9, 49, 50, 42, 16, 16],
    }
    pd.DataFrame(backup).to_csv("cafe_customers.csv", index=False)
    print("Done: cafe_customers.csv is ready.")
```

```python
import pandas as pd
from sklearn.cluster import KMeans

df = pd.read_csv("cafe_customers.csv")
model = KMeans(n_clusters=3, random_state=0)
groups = model.fit_predict(df)
print(groups)
```

A group number (0, 1 or 2) for every customer.

## 6. Let's see the groups
**Before:** just dots. Can you spot groups with your own eyes?

```python
plt.scatter(df["visits"], df["spend"])
plt.xlabel("Visits per month")
plt.ylabel("Average spend (ILS)")
plt.show()
```

**After:** `c=groups` colors each dot by its group. 3 types of customers!

```python
plt.scatter(df["visits"], df["spend"], c=groups)
plt.xlabel("Visits per month")
plt.ylabel("Average spend (ILS)")
plt.show()
```

Rare big spenders, regulars who grab a quick coffee, and the in-between. The model found them without being told.

# ✅ Solutions - try first, peek later!

```python
# Solution 1: 0 hours -> about 47, 10 hours -> about 98.
# 10 hours is a bit outside the examples (max 8), so be careful.
# 0 hours = 47? Nobody in the data studied 0 hours: the model is guessing.
print(grade_model.predict([[0]]))
print(grade_model.predict([[10]]))
```

```python
# Solution 2: banana, but it is a close call!
# Its 3 nearest examples: [150, 8] (apple), [130, 1] and [125, 2] (bananas).
print(fruit_model.predict([[140, 5]]))
```

```python
# Solution 3: with 2 groups, two customer types get merged.
# With 4 groups, one type gets split in two. 3 fits this data best.
for k in [2, 4]:
    kmeans = KMeans(n_clusters=k, random_state=0)
    groups = kmeans.fit_predict(df)
    plt.scatter(df["visits"], df["spend"], c=groups)
    plt.title(f"{k} groups")
    plt.show()
```

```python
# Solution 4
print(grade_model.predict([[2.5], [4.5], [6]]))
```

```python
# Solution 5 (one possible trick): a heavy but yellow fruit.
# Weight numbers (100+) are much bigger than redness numbers (0-10),
# so weight counts more when the model measures "nearest".
print(fruit_model.predict([[175, 1]]))
```
