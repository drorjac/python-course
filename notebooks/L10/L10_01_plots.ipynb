# L10-01: Plots with Matplotlib

Lecture 10, slides: Your first line plot, Always add a title and labels, Two lines on one plot, Bar chart, Scatter plot, Histogram, Try it!, Predict the plot

A list of numbers makes you think. A plot lets you just **see**. Today we turn numbers into pictures with **Matplotlib**, Python's most popular plotting library. The recipe is always the same: import, your data, `plt.plot`, title & labels, `plt.show()`. 📈

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## 1. Your first line plot
My sleep during exam week. `plt.plot(x, y)`: x goes right, y goes up.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
sleep = [8, 7, 7, 6, 5, 4, 9]
plt.plot(days, sleep)
plt.show()
```

Nice line. But... what are we looking at? Nobody knows without a title and labels.

## 2. Always add a title and labels
`marker="o"` draws a dot on each value. Exam on day 6, then sleep!

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
sleep = [8, 7, 7, 6, 5, 4, 9]
plt.plot(days, sleep, marker="o")
plt.title("Exam week: my sleep")
plt.xlabel("Day")
plt.ylabel("Hours of sleep")
plt.show()
```

## 3. Two lines on one plot
`label` + `plt.legend()` = a little key that says which line is which.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
me = [8, 7, 7, 6, 5, 4, 9]
noa = [7, 7, 8, 7, 7, 6, 8]
plt.plot(days, me, label="Me")
plt.plot(days, noa, label="Noa")
plt.ylabel("Hours of sleep")
plt.legend()
plt.show()
```

Noa planned ahead. Be like Noa. 🐍

## 5. Scatter plot: does one thing go with another?
One dot per student (made-up data). Remember these dots: they come back in the machine learning notebook!

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 8]
grades = [52, 58, 60, 68, 72, 78, 88]
plt.scatter(hours, grades)
plt.xlabel("Hours studied")
plt.ylabel("Exam grade")
plt.show()
```

More study, higher grade.

## 6. Histogram: how are values spread?
300 random grades around 72. Each bar counts how many grades fall in a range. `np.random.seed(0)` makes the "random" numbers the same every time, so we all see the same picture.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
grades = np.random.normal(72, 9, 300)
plt.hist(grades, bins=15)
plt.xlabel("Grade")
plt.ylabel("Number of students")
plt.show()
```

## Which plot when?

| Question | Plot | Command |
|---|---|---|
| How does it change over time? | Line | `plt.plot` |
| Which group is bigger? | Bar | `plt.bar` |
| Do two things go together? | Scatter | `plt.scatter` |
| How are values spread? | Histogram | `plt.hist` |
