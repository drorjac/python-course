## 1. `sep`: choose the glue
`print("a", "b")` puts a **space** between the values. `sep` changes that glue.
🔮 **Predict first**, then run.

````python
print("05", "10", "2026", sep="/")
print("ice", "cream", sep="")
print("Dana", "Yossi", "Noa", sep=" & ")
````

## 2. `end`: what comes last
Normally `print` ends with a "new line". `end` replaces it, so the next `print` continues on the **same** line.

````python
print("Loading", end="...")
print("done!")
````

````python
# Solution 5
bill = float(input("Bill in ILS? "))
tip = bill * 0.12
total = bill + tip
print(f"Tip: {tip:.2f} ILS")
print(f"Total: {total:.2f} ILS")
````

````python
# Solution 6
a = float(input("Side a? "))
b = float(input("Side b? "))
c = (a ** 2 + b ** 2) ** 0.5
area = a * b / 2
print(f"Long side: {c}")
print(f"Area: {area}")
````

````python
# Solution 7
shekels = float(input("How many ILS? "))
dollars = shekels / 3.7
print(f"{shekels} ILS = {dollars:.2f} dollars")
````

````python
# Solution 8
age = int(input("Your age? "))
days = age * 365
print(f"That is about {days} days")
print(f"or {days * 24} hours!")
````

````python
# Solution 9: input() gives text, so convert with int() before comparing
grade = int(input("Your grade? "))
if grade >= 56:
    print("Passed! Well done")
else:
    print("Not this time. Try again!")
````

````python
# Solution 10: an if with no else is fine, only small amounts lose the fee
shekels = float(input("How many shekels? "))
rate = 3.7
dollars = shekels / rate
if shekels < 100:
    dollars = (shekels - 5) / rate
print(f"You get {dollars:.2f} dollars")
````

````python
# Solution 11: "Pip" == "pip" is False. Capital letters matter.
word = input("Secret word? ")
if word == "pip":
    print("Come in!")
else:
    print("Wrong word.")
````
