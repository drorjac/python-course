### Solution 1: Pass or fail

`input()` always gives text, so convert it to a number with `int()` before comparing.

```python
grade = int(input("Your grade? "))
if grade >= 56:
    print("Passed! Well done")
else:
    print("Not this time. Try again!")
```

### Solution 2: Shekels to dollars

An `if` with no `else` is fine here -- only small amounts lose the fee.

```python
shekels = float(input("How many shekels? "))
rate = 3.7
dollars = shekels / rate
if shekels < 100:
    dollars = (shekels - 5) / rate
print(f"You get {dollars:.2f} dollars")
```
