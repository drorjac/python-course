## Bug 1: forgot to convert
🔮 **Predict first:** run it and type `20`. What will happen?

```python
age = input("Your age? ")
print(age + 1)
```

**What went wrong:** `age` is the **text** `"20"`, not the number 20.
Text + number? The intern does not know how to do that:
`TypeError: can only concatenate str (not "int") to str` ("concatenate" = glue together).

**Fix:** convert the answer right away with `int(...)`.

```python
age = int(input("Your age? "))
print(age + 1)
```

## Bug 2: text that is not a whole number
🔮 **Predict first:** run it and type `3.5`.

```python
hours = int(input("Hours? "))
```

**What went wrong:** `int()` reads whole numbers only. `"3.5"` has a dot, so:
`ValueError: invalid literal for int() with base 10: '3.5'`.
"Invalid literal" = "I cannot read this as a number".

**Fix:** use `float(...)` for decimals.

```python
hours = float(input("Hours? "))
print(f"You worked {hours} hours")
```

## Bug 3: a number written in words
🔮 **Predict first:** can Python read the word "twenty"?

```python
print(int("twenty"))
```

**What went wrong:** `int()` reads **digits** only. It does not speak English. 🙂
`ValueError: invalid literal for int() with base 10: 'twenty'`.

**Fix:** give it digits.

```python
print(int("20"))
```

## Bug 4: the forgotten colon
🔮 **Predict first:** spot the missing piece before you run it.

```python
age = 20
if age >= 18
    print("Welcome!")
```

**What went wrong:** every `if` line must end with a colon `:`.
Python says exactly that: `SyntaxError: expected ':'`.

**Fix:** add the colon: `if age >= 18:`

```python
age = 20
if age >= 18:
    print("Welcome!")
```

## Bug 5: the missing indent
🔮 **Predict first:** the colon is there now. What else is wrong?

```python
age = 20
if age >= 18:
print("Welcome!")
```

**What went wrong:** the line under the `if` must be pushed right, to show it **belongs to** the `if`:
`IndentationError: expected an indented block after 'if' statement on line 2`.

**Fix:** push `print` 4 spaces to the right.

```python
age = 20
if age >= 18:
    print("Welcome!")
```
