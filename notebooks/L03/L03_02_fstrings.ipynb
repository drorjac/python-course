## 1. Converting between types
| Function | Turns a value into... | Example |
|---|---|---|
| `int()` | a whole number | `int("20")` → `20` |
| `float()` | a decimal number | `float("3.5")` → `3.5` |
| `str()` | text | `str(20)` → `"20"` |

Like changing shekels to dollars: same value, a new form.

```python
print(int("20"))      # text -> whole number
print(float("3.5"))   # text -> decimal number
print(str(20))        # number -> text (looks the same when printed!)
```

## 2. Conversion in action
🔮 **Predict first:** which lines do real math, and which line glues text?

```python
print(int("20") + 1)
print(float("3.5") * 2)
print(str(20) + "20")
```

Lines 1 and 2: now it is real math. Line 3: text glued to text.

## 3. `int()` chops, it does not round
`int(3.9)` throws away everything after the dot. 3.9 hours of sleep is still not 4 hours. 😴

```python
print(int(3.9))
print(float(7))
```

## 4. Fix: convert the answer right away
Read it **inside out**: first `input`, then `int`, then store in `age`.

```python
age = int(input("Your age? "))
print("Next year you will be", age + 1)
```

## 5. f-strings: fill in the blanks
An f-string is a message with **blanks**, like a form letter: *"Hi ___! Your ___ is ready."*
Python fills each blank with a variable's value.
- an `f` right before the opening quote
- each blank is a name in curly braces: `{name}`

```python
name = "Dana"
food = "pizza"
print(f"Hi {name}! Your {food} is ready.")
```

## 6. f-strings can calculate
Numbers work too, no `str()` needed. Any math can go inside `{ }`.

```python
name = "Dana"
age = 20
print(f"{name} is {age} years old")
print(f"Next year: {age + 1}")
```

## 7. Money needs 2 digits: `:.2f`
`:.2f` = "show exactly 2 digits after the dot". It rounds for display.
No more prices like `3.3333333333333335` shekels!

```python
price = 12.5
print(f"Price: {price:.2f} NIS")
print(10 / 3)
print(f"{10 / 3:.2f}")
```

## 8. Example: tip calculator 🧾
A 15% tip means: multiply by 0.15. Try typing `120`.

```python
bill = float(input("Bill in NIS? "))
tip = bill * 0.15
total = bill + tip
print(f"Tip: {tip:.2f} NIS")
print(f"Total: {total:.2f} NIS")
```

## 9. Example: split the pizza bill 🍕
The price can have a dot, so `float`. You can't have 2.5 friends, so `int`.
Try `160` and `3`.

```python
pizza_price = float(input("Price? "))
num_friends = int(input("Friends? "))
each = pizza_price / num_friends
print(f"Each pays {each:.2f} NIS")
```

## 10. Example: the long side of a triangle 📐
Square both short sides, add, take the square root (`** 0.5`).
Try `3` and `4`.

```python
a = float(input("Side a? "))
b = float(input("Side b? "))
c = (a ** 2 + b ** 2) ** 0.5
print(f"Long side: {c}")
```
