## 1. print can show several things
Put commas between the values and `print` shows all of them on one line.
Python puts a **space** between them for you. Text and numbers mix fine here.

```python
print("Coffee costs", 14, "ILS")  # ILS: the shekel's ISO code (you will also see NIS)
print("Dana", "Yossi", "Noa")
```

## 2. Your first conversation
`input()` makes the program **stop and wait** until you type something and press Enter.
Whatever you typed goes into a variable. Like the ATM asking for your PIN.

**Tip:** end the question with a space: `"Your name? "`. Otherwise your answer sticks to the question.

```python
name = input("Your name? ")
print("Hi", name)
```

## 3. Surprise: `input()` always gives text
🔮 **Predict first:** run the cell and type `5`. What will `num + num` show? 10?

```python
num = input("Pick a number: ")
print(type(num))
print(num + num)
```

You typed `5`, but Python sees the **text** `"5"`. And text + text means "glue them together".
Remember this from Lecture 2:

```python
print("5" + "5")   # text + text: glue
print(5 + 5)       # number + number: math
```
