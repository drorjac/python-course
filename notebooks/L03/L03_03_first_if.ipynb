## 1. Yes-or-no questions
| Symbol | Asks... | Example |
|---|---|---|
| `>` `<` | bigger? smaller? | `5 > 3` → `True` |
| `>=` | bigger or equal? | `18 >= 21` → `False` |
| `<=` | smaller or equal? | `45 <= 50` → `True` |
| `==` | equal? | `2 == 2` → `True` |
| `!=` | not equal? | `14 != 12` → `True` |

**Every comparison answers `True` or `False`.** Here are all six, one per line:

```python
print(5 > 3)
print(3 < 5)
print(18 >= 21)
print(45 <= 50)
print(2 == 2)
print(14 != 12)
```

## 3. `=` versus `==`
- `grade = 100` means **store**: put 100 into the box.
- `grade == 100` means **ask**: are they equal?

One `=` tells. Two `==` ask.

```python
grade = 100           # one = : store 100 in the box
print(grade == 100)   # two == : ask "is it 100?"
print(grade == 90)
```

## 4. `if` / `else`: who gets into the party? 🎉
A question with a True/False answer. One action for **yes**, another for **no**.
Only **one** of the two paths runs. `Bye!` is back on the left, so it always runs.

Run it, type `20`. Then run it again and type `16`.

```python
age = int(input("Your age? "))
if age >= 18:
    print("Welcome to the party!")
else:
    print("Sorry, too young")
print("Bye!")
```

## 5. Indentation means "belongs to"
- a colon `:` at the end of the `if` line
- 4 spaces in = belongs to the `if`
- back to the left = always runs

🔮 **Predict first:** then change `34` to `20` and run again. Which line disappears?

```python
temperature = 34
if temperature > 30:
    print("Bring water!")
print("Have a nice day")
```

## 6. Example: free shipping from 200 NIS 📦
Online shops love this trick. Try `150`, then `230`.

```python
cart_total = float(input("Cart total in NIS? "))
if cart_total >= 200:
    print("Free shipping!")
else:
    missing = 200 - cart_total
    print(f"Add {missing:.2f} NIS for free shipping")
```

## 7. Example: even or odd? 🎲
Even = nothing left over when you divide by 2. That is exactly what `%` gives us.

```python
number = int(input("A number? "))
if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
```
