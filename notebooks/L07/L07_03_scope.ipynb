Variables made **inside** a function are **local**: they vanish when the function ends.
Only the `return` value comes out.

## What happens in the function stays in the function

🔮 **Predict first:** the first line works. What about `print(kettle)`?

## Inside can look outside

`city` is **global**: made outside every function. Functions can **read** it. 🏠

```python
city = "Tel Aviv"

def where_am_i():
    print(f"I am in {city}")

where_am_i()
```

## Same name, two different boxes

🔮 **Predict first:** does the coffee break wake up the outside `mood` too?

```python
mood = "sleepy"

def coffee_break():
    mood = "awake"
    print("Inside:", mood)

coffee_break()
print("Outside:", mood)
```

Two boxes with the same label: one in the main program (`"sleepy"`), one inside `coffee_break`
(`"awake"`). The inside box is erased when the function ends.

## Parameters are local too

The function changes **its own** `n`. The outside `n` is still 5.

```python
def square(n):
    n = n * n
    return n

n = 5
result = square(n)
print(result)
print(n)
```

## The golden rule

**Data goes in through parameters and comes out through `return`.**

values ➡️ *parameters* ➡️ ⚙️ your function ➡️ *return* ➡️ answer
