## Surprise: you already use functions

`print`, `len`, `max`, `type`... all functions! A name + `( )` means "do it now".

```python
grades = [85, 92, 58]
print(len(grades))
print(max(grades))
print(type(3.5))
```

## Your very first function

`def` = "define": you **teach** Python a new word. Nothing happens yet!
`say_hi()` = **use** the word (a **call**). Every call runs the steps again.

```python
def say_hi():
    print("Hi there!")
    print("Welcome to class.")

say_hi()
say_hi()
```

## Parameters: fill in the blank

Like a form: "Name: ______". Each call fills in the blank.
`name` is the **parameter** (the blank). `"Dana"` is the **argument** (what you write in it).

```python
def greet(name):
    print(f"Hi {name}, welcome!")

greet("Dana")
greet("Yossi")
```

## Two parameters

A pizza order needs two blanks: the size and the topping. 🍕

```python
def order_pizza(size, topping):
    print(f"One {size} pizza with {topping}, coming up!")

order_pizza("large", "olives")
order_pizza("small", "mushrooms")
```

## Order matters!

🔮 **Predict first:** what happens if we swap the two arguments?

```python
order_pizza("olives", "large")
```

First argument goes to the first parameter, second to the second. The literal intern strikes again!

## A function can hold a loop

Anything you learned so far can go inside a function: `if`, `for`, lists...

```python
def cheer(team, times):
    for i in range(times):
        print(f"Go {team}!")

cheer("Maccabi", 3)
```

## 🔮 Predict the output

Write your guess on paper first. Which line prints first?

```python
def hello(name):
    print("Hi", name)

print("Start")
hello("Noa")
hello("Omer")
print("End")
```

`def` only **teaches** the steps. They run at each **call**, so "Start" comes first.
