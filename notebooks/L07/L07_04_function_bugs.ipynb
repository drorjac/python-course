## Bug 1: calling too early

🔮 **Predict first:** will this work?
(Run this notebook from the top. If you already ran a cell that defines `greet`, Python remembers it
and you won't see the error. Use *Runtime > Restart* to start fresh.)

```python
greet("Dana")

def greet(name):
    print(f"Hi {name}!")
```

### What went wrong
Python reads top to bottom. At line 1, `greet` does not exist yet: you used the word before you taught it.

**Fix:** put the `def` first, then the call.

```python
def greet(name):
    print(f"Hi {name}!")

greet("Dana")
```

## Bug 2: a missing argument

```python
def greet(name):
    print(f"Hi {name}!")

greet()
```

### What went wrong
`greet` needs a name, but the call gave none: an empty blank on the form.

**Fix:** give it a name, `greet("Dana")`, or give the parameter a default: `def greet(name="friend"):`

```python
def greet(name="friend"):
    print(f"Hi {name}!")

greet("Dana")
greet()
```

## Bug 3: the forgotten return

```python
def add_tip(bill):
    total = bill * 1.12

dinner = add_tip(100)
print(dinner + 20)
```

### What went wrong
No `return` means the function hands back `None`. `total` stays inside, `dinner` gets `None`,
and `None + 20` makes no sense.

**Fix:** add `return total` as the last line of the function.

```python
def add_tip(bill):
    total = bill * 1.12
    return total

dinner = add_tip(100)
print(dinner + 20)
```

```python
# Your turn 1: this should print "The bigger one is 8". What does it print instead? Fix bigger().
def bigger(a, b):
    if a > b:
        print(a)
    else:
        print(b)

result = bigger(3, 8)
print("The bigger one is", result)
```

## ✅ Solutions - try first, peek later!

```python
# Solution 1: return the answer instead of printing it
def bigger(a, b):
    if a > b:
        return a
    else:
        return b

result = bigger(3, 8)
print("The bigger one is", result)
```

```python
# Solution 2: order matters - size first, then sugar
make_coffee("small", 2)
```
