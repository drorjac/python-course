### Solution 1: total_price(price, qty)


```python
def total_price(price, qty):
    return price * qty

print(total_price(12, 3))
print(total_price(10, 5))
```

### Solution 2: is_even(n)


```python
def is_even(n):
    return n % 2 == 0

for n in [4, 7, 10, 13]:
    print(n, is_even(n))
```

### Solution 3: average(numbers)


```python
def average(numbers):
    return sum(numbers) / len(numbers)

print(average([88, 41, 72, 56]))
```

### Solution 4: countdown(n)

`range(n, 0, -1)` counts down.

```python
def countdown(n):
    for i in range(n, 0, -1):
        print(i)
    print("Liftoff!")

countdown(5)
```
