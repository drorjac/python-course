## Your first return

`return` sends the answer back. The call `pizza_share(96, 4)` **becomes** `24.0`.

```python
def pizza_share(price, people):
    return price / people

my_part = pizza_share(96, 4)
print(my_part)
print(f"Everyone pays {my_part} ILS")
```

## return vs print: side by side

🔮 **Predict first:** what will `print(a, b)` show? Careful, there are two lines of output!

```python
def double_return(x):
    return x * 2

def double_print(x):
    print(x * 2)

a = double_return(5)
b = double_print(5)
print(a, b)
```

The first `10` was printed **inside** `double_print`. Then `b` got nothing back.
Python calls "nothing" `None`.

## Use the answer like any value

Returned values can be stored, added, printed... A 12% tip: multiply by 1.12. 🧾

```python
def with_tip(bill):
    return bill * 1.12

dinner = with_tip(200)
drinks = with_tip(50)
print(f"Total: {dinner + drinks:.2f} ILS")
```

## return ends the function

At `return` the function stops. For 75, `"Pass"` is returned and the last line never runs.

```python
def passed(grade):
    if grade >= 60:
        return "Pass"
    return "Fail"

print(passed(75))
print(passed(42))
```

## Default values: "the usual"

No sugar given? Use `0`. Defaults go **after** the other parameters. ☕

```python
def make_coffee(size, sugar=0):
    print(f"A {size} coffee with {sugar} sugars")

make_coffee("large")
make_coffee("small", 2)
```

## Write it once, use it on anything

The loop recipe from Lecture 6, now with a name. Any list works: grades, salaries, steps...

```python
def average(numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total / len(numbers)

print(average([85, 92, 58, 74]))
print(average([4200, 8100, 12600]))
```

## The bricks, put together

Big programs are built from small bricks. `report` uses `average` (yours, from the cell above)
and `max` (Python's). A function can call other functions!

```python
def report(name, grades):
    avg = average(grades)
    best = max(grades)
    print(f"{name}: average {avg:.1f}, best {best}")

report("Noa", [85, 92, 58, 74])
report("Omer", [55, 61, 90])
```

## Built-in functions vs. your own

| | Built-in | Your own |
|---|---|---|
| Examples | `print`, `len`, `max` | `greet`, `average` |
| Who wrote it? | The Python team | You, with `def` |
| Ready to use? | Always | After the `def` runs |
| How to call | `len(grades)` | `average(grades)` |

Once defined, yours work just like Python's.
