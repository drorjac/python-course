## The data: a list (Lecture 5!)

```python
days = ["Sun", "Mon", "Tue", "Wed", "Thu"]
coffee = [14, 18, 0, 22, 14]
print(sum(coffee), max(coffee), min(coffee))
print(sum(coffee) / len(coffee))
```

## The simplest chart: 4 lines
`plt.bar(days, coffee)` draws one bar per day.

```python
import matplotlib.pyplot as plt
plt.bar(days, coffee)
plt.title("Coffee spending this week")
plt.show()
```

## The full slide version
Same chart, with colours, the numbers on top of the bars, and a dashed line for the average.

```python
import matplotlib.pyplot as plt
plt.rcParams.update({"font.size": 14})
fig, ax = plt.subplots(figsize=(6, 3.5))
colors = ["#1E3A8A"] * 5          # blue bars (list * 5, like "ha" * 5!)
colors[3] = "#E8554E"             # Wed (the max) in red
ax.bar(days, coffee, color=colors, width=0.6)
for i in range(5):                # loops come in Lecture 6
    ax.text(i, coffee[i] + 0.6, str(coffee[i]), ha="center", fontsize=13)
avg = sum(coffee) / len(coffee)
ax.axhline(avg, color="#2E9E6B", linestyle="--", linewidth=2)
ax.text(2, avg + 0.8, "average 13.6", color="#2E9E6B", ha="center", fontsize=12)
ax.text(3, 25.0, "max", color="#E8554E", ha="center", fontsize=12, fontweight="bold")
ax.text(2, 2.4, "min", color="#B88A00", ha="center", fontsize=12, fontweight="bold")
ax.set_ylabel("ILS")
ax.set_title("Coffee spending this week")
ax.set_ylim(0, 27)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
# fig.savefig("coffee_week.png", dpi=200, bbox_inches="tight")  # remove the # to save it
plt.show()
```
