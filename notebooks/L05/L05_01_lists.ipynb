## A list is a shopping list
Square brackets `[ ]`, commas between the items. `len()` counts how many items are inside.

````python
shopping = ["milk", "eggs", "hummus", "pita"]
print(shopping)
print(len(shopping))
````

## append: add to the end
Like writing a new line at the bottom of your list. The dot means "list, please do this".

````python
shopping = ["milk", "eggs"]
shopping.append("coffee")
shopping.append("bananas")
print(shopping)
````

## insert: cut in line
`insert(pos, item)` puts an item at a position. Positions start at **0** (the front of the line), and everyone behind moves back.
Savta (grandma) always goes first. 👵

````python
queue = ["Noa", "Omer", "Lior"]
queue.insert(1, "Maya")
print(queue)
queue.insert(0, "Savta")
print(queue)
````

## remove and pop: cross things off
- `remove(item)`: remove by **name**
- `pop()`: take the **last** item off and hand it to you

🔮 **Predict first:** what will `last` be?

````python
shopping = ["milk", "eggs", "pita"]
shopping.remove("eggs")
print(shopping)
last = shopping.pop()
print(last)
print(shopping)
````

## in: is it on the list?
`in` gives `True` or `False`. Perfect inside an `if` (Lecture 4!).

````python
shopping = ["milk", "eggs", "tea"]
print("milk" in shopping)
print("pizza" in shopping)
if "pizza" not in shopping:
    print("Add pizza!")
````

## sort: put it in order
Text goes A to Z, numbers go small to big. `reverse=True` puts the big ones first.

````python
names = ["Omer", "Dana", "Lior", "Avi"]
names.sort()
print(names)
grades = [88, 72, 95, 60]
grades.sort(reverse=True)
print(grades)
````

## sum, max, min: my coffee week ☕
How much did I spend on coffee this week (in ILS)? Average = total ÷ how many days.

````python
coffee = [14, 18, 0, 22, 14]
print(sum(coffee))
print(max(coffee), min(coffee))
print(sum(coffee) / len(coffee))
````

## Example: a grade report in 4 lines
f-strings from Lecture 3 + a list. Add a student to the list and nothing else needs to change!

````python
grades = [88, 72, 95, 60, 91]
average = sum(grades) / len(grades)
print(f"Students: {len(grades)}")
print(f"Average: {average:.1f}")
print(f"Best: {max(grades)}, worst: {min(grades)}")
````

## Functions, methods, operators: three kinds of tools

| | Function | Method | Operator |
|---|---|---|---|
| Shape | `name(value)` | `value.name()` | `value op value` |
| What it is | A tool that stands on its own. You hand it the value inside `( )` | An action that **belongs to a type**. The dot means "you, please do this" | A symbol or word that sits **between** two values. No name, no dot, no `( )` |
| Examples so far | `print`, `len`, `input`, `int`, `sum`, `max`, `min` | `append`, `insert`, `remove`, `pop`, `sort`, `lower`, `strip`, `count` | `+`, `//`, `==`, `>=`, `and`, `or`, `not`, `in` |

🔭 Coming up: in **Lecture 7** you write your own functions, and in **Lecture 9** you build your own types (classes) with methods of their own.

````python
names = ["Omer", "Dana"]
print(len(names))       # function: names goes inside ( )
names.append("Avi")     # method: names comes before the dot
print(names)
print("Avi" in names)   # operator: in sits between two values
````

## Each type has its own methods
A list can `append`. A string can't. Try it in a new cell:
```python
"hi".append("!")
```
You get: `AttributeError: 'str' object has no attribute 'append'`. Python means: "strings don't know how to do that".

## A string method: replace
`text.replace(old, new)`: every `old` becomes `new`. Good for dropping dashes, dropping spaces, fixing a word everywhere.
It gives back a **new** string; the original stays the same.

````python
phone = "050-123-4567"
print(phone.replace("-", ""))
msg = "I love Java"
print(msg.replace("Java", "Python"))
print("a b c".replace(" ", ""))
print(phone)  # still with dashes
````

## 🪄 Alohomora! ... nothing happens?
The door only opens for exactly `"alohomora"`. Capitals or spaces, and nothing happens.
*(On the slide the spell comes from `input()`. Here it is fixed in the code, so you can change it and rerun.)*
