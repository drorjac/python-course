## A list is a shopping list
Square brackets `[ ]`, commas between the items. `len()` counts how many items are inside.

```python
shopping = ["milk", "eggs", "hummus", "pita"]
print(shopping)
print(len(shopping))
```

## append: add to the end
Like writing a new line at the bottom of your list. The dot means "list, please do this".

```python
shopping = ["milk", "eggs"]
shopping.append("coffee")
shopping.append("bananas")
print(shopping)
```

## insert: cut in line
`insert(pos, item)` puts an item at a position. Positions start at **0** (the front of the line), and everyone behind moves back.
Savta (grandma) always goes first. 👵

```python
queue = ["Noa", "Omer", "Lior"]
queue.insert(1, "Maya")
print(queue)
queue.insert(0, "Savta")
print(queue)
```

## remove and pop: cross things off
- `remove(item)`: remove by **name**
- `pop()`: take the **last** item off and hand it to you

🔮 **Predict first:** what will `last` be?

```python
shopping = ["milk", "eggs", "pita"]
shopping.remove("eggs")
print(shopping)
last = shopping.pop()
print(last)
print(shopping)
```

## in: is it on the list?
`in` gives `True` or `False`. Perfect inside an `if` (Lecture 4!).

```python
shopping = ["milk", "eggs", "tea"]
print("milk" in shopping)
print("pizza" in shopping)
if "pizza" not in shopping:
    print("Add pizza!")
```

## sort: put it in order
Text goes A to Z, numbers go small to big. `reverse=True` puts the big ones first.

```python
names = ["Omer", "Dana", "Lior", "Avi"]
names.sort()
print(names)
grades = [88, 72, 95, 60]
grades.sort(reverse=True)
print(grades)
```

## sum, max, min: my coffee week ☕
How much did I spend on coffee this week (in ILS)? Average = total ÷ how many days.

```python
coffee = [14, 18, 0, 22, 14]
print(sum(coffee))
print(max(coffee), min(coffee))
print(sum(coffee) / len(coffee))
```

## Example: a grade report in 4 lines
f-strings from Lecture 3 + a list. Add a student to the list and nothing else needs to change!

```python
grades = [88, 72, 95, 60, 91]
average = sum(grades) / len(grades)
print(f"Students: {len(grades)}")
print(f"Average: {average:.1f}")
print(f"Best: {max(grades)}, worst: {min(grades)}")
```
