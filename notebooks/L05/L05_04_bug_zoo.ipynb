## Bug 1: there is no seat number 3
🔮 Predict first: what will the intern say?

```python
queue = ["Noa", "Omer", "Lior"]
print(queue[3])
```

### What went wrong
**Off by one.** 3 people sit in seats 0, 1, 2. There is no seat 3, so Python says `IndexError: list index out of range`.
**Fix:** `queue[2]`, or `queue[-1]` for the last one.

```python
queue = ["Noa", "Omer", "Lior"]
print(queue[2])
print(queue[-1])
```

## Bug 2: removing something that isn't there

```python
shopping = ["milk", "eggs"]
shopping.remove("pizza")
```

### What went wrong
You can't cross off pizza if it's not on the list: `ValueError: list.remove(x): x not in list`.
**Fix:** check first with `if "pizza" in shopping:`.

```python
shopping = ["milk", "eggs"]
if "pizza" in shopping:
    shopping.remove("pizza")
else:
    print("No pizza on the list, nothing to remove")
print(shopping)
```

## Bug 3: asking a set for position 0

```python
group = {"Dana", "Omer"}
print(group[0])
```

### What went wrong
`TypeError: 'set' object is not subscriptable`. "Not subscriptable" = "you can't use `[ ]` on this".
A set has no order, so it has no positions.
**Fix:** sets have no positions. Need a first item? Use a list.

```python
group = ["Dana", "Omer"]
print(group[0])
```
