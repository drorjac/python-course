## Every item has a position number

| | "Noa" | "Omer" | "Lior" | "Avi" |
|---|---|---|---|---|
| **index** | 0 | 1 | 2 | 3 |
| **from the end** | -4 | -3 | -2 | -1 |

4 people sit in positions 0, 1, 2, 3. Negative numbers count from the back: **-1** is the last.

## Grab one item: `[ ]`

```python
queue = ["Noa", "Omer", "Lior", "Avi"]
print(queue[0])
print(queue[1])
print(queue[3])
```

Strings work the same way! 🔮 Which letter is at position 3?

```python
word = "PYTHON"
print(word[0])
print(word[3])
```

## Negative indexes: from the back
`[-1]` is the last one, always. No need to know the length.

```python
queue = ["Noa", "Omer", "Lior", "Avi"]
print(queue[-1])
print(queue[-2])
word = "PYTHON"
print(word[-1])
```

## Slicing: a piece of the list
`[start:stop]`: from `start`, up to **but not including** `stop`.
`week[0:5]` is the Israeli work week, Sun to Thu.

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[1:4])
print(week[0:5])
```

## Leave out start or stop
- `[:3]`: from the start
- `[5:]`: to the end
- `[-2:]`: the last two 🏖️

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[:3])
print(week[5:])
print(week[-2:])
```

## Slicing text 🍫
Slices work on strings too.

```python
word = "chocolate"
print(word[:5])
print(word[5:])
```

Cut a name out of an email address, or grab its ending:

```python
email = "dana.levi@mail.tau.ac.il"
print(email[:9])
print(email[-6:])
```

## Steps: `[start:stop:step]`
Step 2: every second item. Step -1: backwards!
🔮 **Predict first:** what is "stressed" backwards? (Hint: exam week 🍰)

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[::2])
print("stressed"[::-1])
print("level"[::-1])
```

## Lists can change. Strings can't.
You can swap one song in a list:

```python
songs = ["Toxic", "Halo"]
songs[1] = "Yellow"
print(songs)
```

But try that on a string and the intern complains. This cell is **supposed** to show a red error:

```python
word = "cat"
word[0] = "b"
```

💡 Build a **new** string instead: `"b" + word[1:]` gives `"bat"`.

```python
word = "cat"
new_word = "b" + word[1:]
print(new_word)
```

## Elements made of elements 🧩
A list can hold things that have their **own** elements: strings, or even other lists.
Read `houses[0][1]` left to right: **outer** index first (which house), then **inner** index (which letter).
🔮 **Predict first:** what is `houses[3][-1]`?

```python
houses = ["Gryffindor", "Hufflepuff", "Ravenclaw", "Slytherin"]
print(houses[0])
print(houses[0][1])
print(houses[3][-1])
print(len(houses[1]))
```

## The Sorting Hat 🎩: one empty list per house
A list of 4 elements, and each element is an **empty list**. With N houses the positions go from `hogwarts[0]` to `hogwarts[N-1]`:

| `[0]` | `[1]` | `[2]` | `[3]` |
|---|---|---|---|
| Gryffindor | Hufflepuff | Ravenclaw | Slytherin |

```python
students = ["Harry", "Hermione", "Cedric", "Luna", "Draco"]
hogwarts = [[], [], [], []]
print(hogwarts, len(hogwarts))
```

Now the Hat sorts. `hogwarts[0]` is a list, so it has `.append()`:

```python
hogwarts[0].append(students[0])
hogwarts[0].append(students[1])
hogwarts[1].append(students[2])
hogwarts[2].append(students[3])
hogwarts[3].append(students[4])
print(hogwarts)
print(hogwarts[3][0][0])  # house 3, student 0, letter 0
```

💡 5 students, 5 lines. 300 students? Next lecture: **loops** do the sorting for us. (Lecture 8: lists of lists as tables.)

## 🔮 Predict the output
Careful with the stop number!

```python
nums = [10, 20, 30, 40, 50]
print(nums[1:3])
print(nums[-1])
print(nums[::-1])
```

💡 **Why?** Stop 3 is not included, so `40` is left out.
