Every item in a list (and every letter in a string) sits in a numbered seat. 🐍 Pip's tip: the literal intern starts counting at **0**, not 1 (computers are weird).

## Every item has a position number

| | "Noa" | "Omer" | "Lior" | "Avi" |
|---|---|---|---|---|
| **index** | 0 | 1 | 2 | 3 |
| **from the end** | -4 | -3 | -2 | -1 |

4 people sit in positions 0, 1, 2, 3. Negative numbers count from the back: **-1** is the last.

## Grab one item: `[ ]`

```python
queue = ["Noa", "Omer", "Lior", "Avi"]
print(queue[0])
print(queue[1])
print(queue[3])
```

Strings work the same way! 🔮 Which letter is at position 3?

```python
word = "PYTHON"
print(word[0])
print(word[3])
```

## Negative indexes: from the back
`[-1]` is the last one, always. No need to know the length.

```python
queue = ["Noa", "Omer", "Lior", "Avi"]
print(queue[-1])
print(queue[-2])
word = "PYTHON"
print(word[-1])
```

## Slicing: a piece of the list
`[start:stop]`: from `start`, up to **but not including** `stop`.
`week[0:5]` is the Israeli work week, Sun to Thu.

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[1:4])
print(week[0:5])
```

## Leave out start or stop
- `[:3]`: from the start
- `[5:]`: to the end
- `[-2:]`: the last two 🏖️

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[:3])
print(week[5:])
print(week[-2:])
```

## Slicing text 🍫
Slices work on strings too.

```python
word = "chocolate"
print(word[:5])
print(word[5:])
```

Cut a name out of an email address, or grab its ending:

```python
email = "dana.levi@mail.tau.ac.il"
print(email[:9])
print(email[-6:])
```

## Steps: `[start:stop:step]`
Step 2: every second item. Step -1: backwards!
🔮 **Predict first:** what is "stressed" backwards? (Hint: exam week 🍰)

```python
week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(week[::2])
print("stressed"[::-1])
print("level"[::-1])
```

## Lists can change. Strings can't.
You can swap one song in a list:

```python
songs = ["Toxic", "Halo"]
songs[1] = "Yellow"
print(songs)
```

But try that on a string and the intern complains. This cell is **supposed** to show a red error:

```python
word = "cat"
word[0] = "b"
```

💡 Build a **new** string instead: `"b" + word[1:]` gives `"bat"`.

```python
word = "cat"
new_word = "b" + word[1:]
print(new_word)
```

## 🔮 Predict the output
Careful with the stop number!

```python
nums = [10, 20, 30, 40, 50]
print(nums[1:3])
print(nums[-1])
print(nums[::-1])
```

💡 **Why?** Stop 3 is not included, so `40` is left out.
