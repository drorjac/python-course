## A set: nobody is in twice
Curly brackets `{ }`. Duplicates disappear by themselves.

```python
group = {"Dana", "Omer", "Dana", "Lior"}
print(group)
print(len(group))
```

> 🔀 **Heads-up:** a set has no order. When you run this, the names may come out in a different order than on the slide (or than your neighbour's screen). That's fine: it's the same set.

## Drop the duplicates
Dana clicked "sign up" 3 times. How many **real** students are there?
`set(...)` turns a list into a set: an instant duplicate remover.

```python
signups = ["Dana", "Omer", "Dana", "Lior", "Dana", "Maya"]
students = set(signups)
print(len(signups))
print(len(students))
```

## add: put one more in
Sets use `add`, lists use `append`. Adding `"cat"` again does nothing: no error, no copy.

🔮 **Predict first:** how many stickers at the end?

```python
stickers = {"cat", "pizza"}
stickers.add("rocket")
stickers.add("cat")
print(stickers)
print(len(stickers))
```

## Two friends, two playlists 🎧
Dana and Omer are planning a party. Which artists do they **share**? What goes on the **party playlist**?

| Only Dana | Both | Only Omer |
|---|---|---|
| Coldplay, Noa Kirel | Adele, Queen | Eminem, Omer Adam |

### `&`: what both like (intersection)
The middle of the two circles. **Intersection** = "in both".

```python
dana = {"Adele", "Coldplay", "Noa Kirel", "Queen"}
omer = {"Queen", "Eminem", "Adele", "Omer Adam"}
print(dana & omer)
```

### `|`: everything (union)
**Union** = "in at least one". The party playlist, with Adele only once.
`|` is the "pipe" key, usually above **Enter**. (Run the cell above first, so `dana` and `omer` exist.)

```python
party = dana | omer
print(party)
print(len(party))
```

### `-`: only mine (difference)
**Difference** = "in mine, not in yours". Great for: songs to recommend to your friend.

```python
print(dana - omer)
print(omer - dana)
```

> 🔀 **Heads-up:** a set has no order. When you run this, the names may come out in a different order than on the slide (or than your neighbour's screen). That's fine: it's the same set.

## List or set?

| | List `[ ]` | Set `{ }` |
|---|---|---|
| Keeps order | ✅ | ❌ |
| Allows duplicates | ✅ | ❌ |
| Grab item by position | ✅ | ❌ |
| Compare two groups (`&`, `\|`, `-`) | ❌ | ✅ |

- Playlist → list
- Who signed up → set

## 🔮 Predict the output
Guess all three lines first!

```python
a = {1, 2, 3, 3, 3}
b = {3, 4}
print(len(a))
print(a | b)
print(a & b)
```

💡 **Why?** The extra 3s vanish. Only 3 is in both.
