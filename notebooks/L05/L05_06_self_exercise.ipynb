### Solution 1: What should Yoni watch?

`&` is shared, `-` is mine-only, `|` is all. Sets have no order, so sort before printing so the output looks the same every time.

````python
noa = {"Fauda", "Friends", "Shtisel", "Lupin"}
yoni = {"Friends", "Dark", "Fauda", "Narcos"}
print(f"They share {len(noa & yoni)} shows")
for_yoni = list(noa - yoni)
for_yoni.sort()
print(f"For Yoni: {for_yoni}")
print(f"In total: {len(noa | yoni)} shows")
````

### Solution 2: The anagram detective

Clean (lowercase, no spaces), turn into a list, sort, compare. Pairs 1, 2, 4 are anagrams, 3 is not.
A set says `True` for pair 3 too: it forgets **how many** of each letter there are.

````python
a = "Dormitory".lower().replace(" ", "")
b = "dirty room".lower().replace(" ", "")
la, lb = list(a), list(b)
la.sort()
lb.sort()
print(la == lb)

x, y = list("aab"), list("abb")
x.sort()
y.sort()
print(x == y)
print(set("aab") == set("abb"))
````

### Solution 3: Mirror, mirror

One letter and the empty string read the same both ways: `True`. Slicing works on lists too.
For the sentence, drop spaces and capitals first.

````python
print("a" == "a"[::-1])
print("" == ""[::-1])
nums = [1, 2, 3, 2, 1]
print(nums == nums[::-1])
s = "Was it a car or a cat I saw".lower().replace(" ", "")
print(s == s[::-1])
````

### Solution 4: Spell check

`print` and `len` are functions; `strip`, `lower`, `replace` are methods. `.strip()` drops spaces only at the ends; `.replace(" ", "")` drops all of them.

````python
spell = "  AloHoMora  "
print(spell.strip())
print(spell.lower())
print(spell.strip().lower() == "alohomora")
print(spell.replace(" ", "").lower())
print(len(spell), len(spell.strip()))
````
