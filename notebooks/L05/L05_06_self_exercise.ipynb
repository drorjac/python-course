### Solution 1: What should Yoni watch?

`&` is shared, `-` is mine-only, `|` is all. Sets have no order, so sort before printing so the output looks the same every time.

```python
noa = {"Fauda", "Friends", "Shtisel", "Lupin"}
yoni = {"Friends", "Dark", "Fauda", "Narcos"}
print(f"They share {len(noa & yoni)} shows")
for_yoni = list(noa - yoni)
for_yoni.sort()
print(f"For Yoni: {for_yoni}")
print(f"In total: {len(noa | yoni)} shows")
```
