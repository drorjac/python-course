At the supermarket you keep a running total in your head: start at 0, add each item as it goes into the cart. 🛒
That's a **loop recipe**. 🐍 Pip has five of them, and they solve a huge number of problems.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## Recipe 1: the total (the accumulator)
Start at 0, add each price. `sum()` does the same: now you know what it does inside!

```python
coffee = [12, 14, 0, 18, 12]  # ILS
total = 0
for price in coffee:
    total = total + price
print(f"Total: {total} ILS")
print(sum(coffee))
```

## Recipe 2: the average
Average = total ÷ how many. Here: 475 points over 6 exams.

```python
grades = [85, 92, 58, 74, 100, 66]
total = 0
for grade in grades:
    total = total + grade
average = total / len(grades)
print(f"Average: {average:.1f}")
```

## Recipe 3: counting
A counter is an accumulator that adds **1**, only **if**...

```python
grades = [85, 92, 58, 74, 100, 66, 45]
passed = 0
for grade in grades:
    if grade >= 60:
        passed = passed + 1
print(f"{passed} of {len(grades)} passed")
```

## Recipe 4: the maximum 👟
Start with the first item as the "best so far". Every time you see a bigger one, remember it.
The "tallest student" algorithm from Lecture 1!

```python
steps = [4200, 8100, 12500, 3900,
         9800, 15200, 7000]
best = steps[0]
for day_steps in steps:
    if day_steps > best:
        best = day_steps
print(f"Best day: {best} steps")
print(max(steps))
```

## Recipe 5: search 🕵️
Start with `found = False`. If you find a match: `True`, and `break` (no need to keep looking).

```python
guests = ["Noa", "Omer", "Dana", "Yael"]
found = False
for guest in guests:
    if guest == "Dana":
        found = True
        break
print(f"Dana invited? {found}")
```

Shortcut from Lecture 5: `in` does this loop for you.

```python
guests = ["Noa", "Omer", "Dana", "Yael"]
print("Dana" in guests)
```

## All together: my coffee week ☕
One loop, two recipes at once: a total **and** a counter.

```python
week = [12, 14, 0, 18, 12, 9, 16]
total = 0
big_days = 0
for price in week:
    total = total + price
    if price > 13:
        big_days = big_days + 1
print(f"Week total: {total} ILS")
print(f"Average: {total / len(week):.2f} ILS")
print(f"Days over 13 ILS: {big_days}")
```

## Cheat sheet: four loop recipes 📝

| Recipe | Before the loop | Inside the loop |
|---|---|---|
| Total | `total = 0` | add the item |
| Count | `count = 0` | add 1, if... |
| Max | `best = first item` | bigger? remember it |
| Search | `found = False` | match? True, `break` |

**Set up before, update inside, use after.**

## 🔮 Predict the output
The counter recipe, on a string. 🍌

```python
word = "banana"
count = 0
for letter in word:
    if letter == "a":
        count = count + 1
print(count)
```

💡 **Why?** b-**a**-n-**a**-n-**a**: the counter recipe on a string.
