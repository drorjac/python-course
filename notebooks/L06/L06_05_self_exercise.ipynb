### Solution 1: Saving for a festival

The number of weeks is unknown in advance, so `while` fits better than `for`.

```python
savings = 150
weeks = 0
while savings < 450:
    savings = savings + 75
    weeks = weeks + 1
    print(f"Week {weeks}: {savings} ILS")
print(f"Ticket after {weeks} weeks!")
```

### Solution 2: Password checker

`in` works on text too: `"5" in "0123456789"` is `True`.

```python
password = input("Password? ")
digits = 0
for letter in password:
    if letter in "0123456789":
        digits = digits + 1
print(f"Digits: {digits}")
if len(password) >= 8 and digits >= 2:
    print("Strong password!")
else:
    print("Weak password")
```
