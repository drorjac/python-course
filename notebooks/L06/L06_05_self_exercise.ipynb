### Solution 1: Saving for a festival

The number of weeks is unknown in advance, so `while` fits better than `for`.

````python
savings = 150
weeks = 0
while savings < 450:
    savings = savings + 75
    weeks = weeks + 1
    print(f"Week {weeks}: {savings} ILS")
print(f"Ticket after {weeks} weeks!")
````

### Solution 2: Password checker

`in` works on text too: `"5" in "0123456789"` is `True`.

````python
password = input("Password? ")
digits = 0
for letter in password:
    if letter in "0123456789":
        digits = digits + 1
print(f"Digits: {digits}")
if len(password) >= 8 and digits >= 2:
    print("Strong password!")
else:
    print("Weak password")
````

### Solution 3: Half is enough

The pairs are (0,-1), (1,-2), (2,-3), (3,-4)... after the middle, the same pairs come back in reverse. `7 // 2` is 3: positions 0, 1, 2. The middle letter is its own partner, so it never needs checking.

````python
for word in ["racecar", "abba", "abca"]:
    is_pal = True
    for i in range(len(word) // 2):
        if word[i] != word[-1 - i]:
            is_pal = False
    print(word, is_pal)
````

### Solution 4: How many tries?

The counter recipe, inside a `while` loop. With `Abracadabra` then `ALO HO MORA`: 2 tries.

````python
spell = input("Say the spell: ")
tries = 1
while spell.lower().replace(" ", "") != "alohomora":
    print("Nothing happens...")
    spell = input("Say the spell: ")
    tries = tries + 1
print(f"The door opens! Tries: {tries}")
````

### Solution 5: Bug hunt

`in` only checks **that** a letter appears, not **how many times**, and never looks at extra letters in `b`.
`"aab"` and `"ab"` give `True`, and so do `"ab"` and `"abc"`. Fix: compare lengths and counts.

````python
for a, b in [("listen", "silent"), ("aab", "ab"), ("ab", "abc")]:
    same = len(a) == len(b)
    for letter in a:
        if a.count(letter) != b.count(letter):
            same = False
    print(a, b, same)
````
