## Example: rocket countdown 🚀
The two indented lines are the **body**. They repeat. The last line is not indented, so it runs once.

```python
count = 3
while count > 0:
    print(count)
    count = count - 1
print("Liftoff!")
```

## Slow motion: what the intern does

| count | `count > 0`? | prints |
|---|---|---|
| 3 | True | 3 |
| 2 | True | 2 |
| 1 | True | 1 |
| 0 | **False** | Liftoff! |

False means: leave the loop. Want to *see* the check happen? Run this version, it prints the question each round:

```python
count = 3
while count > 0:
    print(f"count is {count}, count > 0 is {count > 0}")
    count = count - 1
print(f"count is {count}, count > 0 is {count > 0}")
print("Liftoff!")
```

## How many coffees can I afford? ☕
We don't know in advance how many rounds. `while` finds out.
🔮 **Predict first:** 50 ILS, 12 ILS per coffee. How many cups, and how much money is left?

```python
money = 50
coffee_price = 12
cups = 0
while money >= coffee_price:
    money = money - coffee_price
    cups = cups + 1
print(f"I can buy {cups} coffees")
print(f"Money left: {money} ILS")
```

## Keep asking until it's right 🔒
The loop keeps asking as long as the password is wrong. Run it and try a few wrong passwords first.
(The secret is `falafel123`. Shhh.)

```python
password = input("Password? ")
while password != "falafel123":
    print("Wrong! Try again.")
    password = input("Password? ")
print("Welcome in!")
```

## ⚠️ Careful: the loop that never ends
In the code below, `count` never changes, so `count > 0` stays True **forever**. It would print `3` again and again.

That's why it is **commented out**. If you want to see it (brave!), remove the `#` signs and run it, then
**stop it** with the cell's **stop button** (the square ⏹ next to the cell, or *Runtime → Interrupt execution* in Colab).

```python
# count = 3
# while count > 0:
#     print(count)
print("Infinite loop is switched off. Phew!")
```

The fix: change `count` inside the loop, so one day the condition becomes False.

```python
count = 3
while count > 0:
    print(count)
    count = count - 1
```

## 🔮 Predict the output
How many times does the loop run? What is printed?

```python
n = 1
while n < 20:
    n = n * 2
print(n)
```

💡 **Why?** `n` goes 1, 2, 4, 8, 16, 32. 32 is not less than 20, so the loop stops. `print` is not indented: it runs once.
