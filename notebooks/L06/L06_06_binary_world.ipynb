## Bits and bytes

A **bit** is one 0 or 1. A **byte** is 8 bits in a row. Each bit doubles the options, so a byte can hold 2 ** 8 = **256** different values: 0 to 255.

```python
print(2 ** 8)
print(2 ** 1, 2 ** 2, 2 ** 3)
```

## Everything is numbers, even letters

`ord` turns a letter into its number, `chr` turns a number back into a letter.
Your WhatsApp message is a list of numbers, stored as bits. 📱

```python
print(ord("A"))
print(chr(65))
for letter in "Hi":
    print(letter, ord(letter))
```

## 13 in binary: keep dividing by 2

| number | remainder |
|---|---|
| 13 // 2 = 6 | 1 |
| 6 // 2 = 3 | 0 |
| 3 // 2 = 1 | 1 |
| 1 // 2 = 0 | 1 |

Read the remainders from the **bottom up**: `1101`. Check: 8 + 4 + 0 + 1 = 13 ✅

## The same recipe as a while loop

`n % 2` is the remainder: our next bit. `str(...) + bits` puts it in **front**, so we read bottom up. `n // 2` divides and keeps going.
`bin(13)` agrees (`0b` just means "binary").

```python
n = 13
bits = ""
while n > 0:
    bits = str(n % 2) + bits
    n = n // 2
print(bits)
print(bin(13))
```

## and, or, not with 0s and 1s

| a | b | a and b | a or b |
|---|---|---|---|
| 0 | 0 | 0 | 0 |
| 0 | 1 | 0 | 1 |
| 1 | 0 | 0 | 1 |
| 1 | 1 | 1 | 1 |

**not** flips: not 0 is 1, not 1 is 0. In Python, **True** is 1 and **False** is 0: same tables, two spellings.

```python
print(True and False)
print(True or False)
print(not True)
print(int(True), int(False))
```
