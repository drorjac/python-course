## Wish 300 friends a happy new year?
**Without a loop:** one line per friend... and 297 more lines. 😱

```python
print("Happy new year!")
print("Happy new year!")
print("Happy new year!")
```

**With a loop:** two lines. Done. (Get ready to scroll!)

```python
for i in range(300):
    print("Happy new year!")
```

## Example: play the playlist 🎧
`song` is a new variable. Python fills it for you: round 1 it is `"Shape of You"`, round 2 `"Despacito"`, round 3 `"Toxic"`. No items left: the loop ends.

```python
playlist = ["Shape of You", "Despacito", "Toxic"]
for song in playlist:
    print(f"Now playing: {song}")
print("Playlist over!")
```

## A string is a list of letters
A `for` loop works on anything you can index: lists **and** strings.

```python
for letter in "Pizza":
    print(letter)
```

## Repeat N times with `range` 🥳
`range(3)` gives 0, 1, 2: three rounds. You don't have to use `i`.

```python
for i in range(3):
    print("Hip hip hooray!")
```

## The many faces of `range`
`range(start, stop, step)`: start is in, stop is **never** in. Same rule as slicing (Lecture 5)!
`list(...)` lets us peek at the numbers.
🔮 **Predict first:** write down all four lines.

```python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -2)))
```

## Example: the 7 times table
This time we **use** `i`: it is 1, then 2, ..., then 5.

```python
for i in range(1, 6):
    print(f"7 x {i} = {7 * i}")
```

## while or for?

| Use `for` when... | Use `while` when... |
|---|---|
| you have a list or a string | you wait for something to happen |
| you know how many rounds | you don't know how many rounds |
| every song, every grade, 10 times | right password, money runs out |

## Emergency exit: `break` 🦈
`break` leaves the loop **right now**. "Despacito" never plays.

```python
playlist = ["Toxic", "Baby Shark", "Despacito"]
for song in playlist:
    if song == "Baby Shark":
        print("Nope. Stopping the music!")
        break
    print(f"Now playing: {song}")
```

## 🔮 Predict the output
Start, stop, step...

```python
for i in range(2, 10, 3):
    print(i)
```

💡 **Why?** Start at 2, jump by 3. The next one, 11, is past 10.
