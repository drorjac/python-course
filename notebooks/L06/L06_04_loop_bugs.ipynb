Loops follow the same rules as `if` (Lecture 4): a **colon**, then an **indent**. 🐍 Pip collected four loop bugs for you. Three shout a red error. The fourth one is sneaky: **no error at all**, just a wrong answer.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output! The first three bug cells *should* show an error. Then read "What went wrong" and run the fixed version.

## Bug 1: the missing colon

```python
for i in range(3)
    print(i)
```

### What went wrong
`SyntaxError: expected ':'`. Every `for` line (like `if` and `while`) ends with a colon.
**Fix:** `for i in range(3):`

```python
for i in range(3):
    print(i)
```

## Bug 2: the missing indent

```python
for i in range(3):
print(i)
```

### What went wrong
The intern doesn't know which lines belong to the loop: `IndentationError: expected an indented block`.
**Fix:** indent the body with 4 spaces (or press **Tab**).

```python
for i in range(3):
    print(i)
```

## Bug 3: looping over a number

```python
for i in 5:
    print(i)
```

### What went wrong
**A number is not a list.** "Iterable" = something you can walk through, item by item (a list, a string, a `range`). `5` is just one number.
**Fix:** `for i in range(5):`

```python
for i in range(5):
    print(i)
```

## Bug 4: the silent bug 🤫
🔮 **Predict first:** we expect the total to be 56. What does it really print?

```python
coffee = [12, 14, 0, 18, 12]
for price in coffee:
    total = 0
    total = total + price
print(total)
```

### What went wrong
We expected 56. No error message! `total = 0` runs in **every** round, so the total starts over each time. In the end it only holds the last price, 12.
**Fix:** move `total = 0` **above** the loop (set up before, update inside, use after).

```python
coffee = [12, 14, 0, 18, 12]
total = 0
for price in coffee:
    total = total + price
print(total)
```

### Exercise 1
We want to count the passing grades (60 and up). There are 3, but the code says 1. Why?

```python
grades = [85, 40, 92, 58, 70]
for grade in grades:
    passed = 0
    if grade >= 60:
        passed = passed + 1
print(f"Passed: {passed}")
```

We want the 7 times table from 1 to **10**, but it stops at 9. Why?

```python
for i in range(1, 10):
    print(f"7 x {i} = {7 * i}")
```

# ✅ Solutions - try first, peek later!

**Exercise 1:** `passed = 0` is inside the loop, so the counter resets every round. Move it above the loop.

```python
grades = [85, 40, 92, 58, 70]
passed = 0
for grade in grades:
    if grade >= 60:
        passed = passed + 1
print(f"Passed: {passed}")
```

**Exercise 2:** the stop is never in. To reach 10, the stop must be 11.

```python
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
```
