## Confidently wrong

You asked for a median function. The AI answers: *"Here is a median function!"*
🔮 **Predict first:** what is the median of 5, 1, 9? What will this code print?

```python
def median(values):
    middle = len(values) // 2
    return values[middle]

print(median([5, 1, 9]))
```

The median of 5, 1, 9 is **5**! The code forgot to **sort**. It looks right, and it is wrong.

## Always test AI code

Use cases you can solve **by hand**, and add tricky ones: an even count, a single item.

```python
print(median([5, 1, 9]), "expected 5")
print(median([4, 1, 3, 2]), "expected 2.5")
print(median([7]), "expected 7")
```

### Make the tests answer True or False

Comparing with `==` lets Python do the checking for you. Every line should say `True`.

```python
print(median([5, 1, 9]) == 5)
print(median([4, 1, 3, 2]) == 2.5)
print(median([7]) == 7)
```

### The fixed median passes every test

This is the median function from the lecture: sort first, and handle an even count.

```python
def median(values):
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:     # odd count
        return ordered[middle]
    else:                         # even count
        return (ordered[middle - 1] + ordered[middle]) / 2
```

```python
print(median([5, 1, 9]) == 5)
print(median([4, 1, 3, 2]) == 2.5)
print(median([7]) == 7)
```

## 🔮 Predict the output: the AI wrote this

The rule: **passing grade is 60 and up**. Two lines will print. What are they?

```python
def is_passing(grade):
    if grade > 60:
        return True
    else:
        return False
print(is_passing(75))
print(is_passing(60))
```

A 60 should pass! The fix: `>=`. **Always test the edge**: just below, exactly on, just above.

```python
def is_passing(grade):
    if grade >= 60:
        return True
    else:
        return False
print(is_passing(59), "expected False")
print(is_passing(60), "expected True")
print(is_passing(61), "expected True")
```

## ✅ Solutions - try first, peek later!

**Solution 1** (one good prompt):

> I am a beginner in Python. Write a function `most_common(votes)` that gets a list of words and
> returns the word that appears most often. Example: `["pizza", "sushi", "pizza"]` ➡️ `"pizza"`.
> If there is a tie, return the one that appeared first in the list.
> Use only `for` loops, `if` and dictionaries. Explain each line simply.

```python
# Solution 2a: one answer an AI might give (only things we learned)
def most_common(votes):
    counts = {}
    for v in votes:
        counts[v] = counts.get(v, 0) + 1
    best = None
    best_count = 0
    for item, count in counts.items():
        if count > best_count:
            best = item
            best_count = count
    return best
```

```python
# Solution 2b: with this version a tie goes to the item seen first
print(most_common(["pizza", "sushi", "pizza"]), "expected pizza")
print(most_common(["tea"]), "expected tea")
print(most_common(["tea", "coffee", "coffee", "tea"]), "expected tea (seen first)")
```
