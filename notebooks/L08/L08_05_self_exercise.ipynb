### Solution 2: Word count


```python
words = ['cat', 'dog', 'cat', 'bird', 'dog', 'cat']
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

### Solution 3: Pass or fail, for everyone


```python
names = ['Noa', 'Tom', 'Omer', 'Lior']
scores = [88, 41, 72, 56]
results = dict(zip(names, scores))
for name, score in results.items():
    if score >= 56:
        print(name, "Passed! Well done")
    else:
        print(name, "Not this time. Try again!")
```
