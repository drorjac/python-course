## A logic bug: no error, wrong answer

We expected 60. 🔮 What do we get?

```python
def total_price(prices):
    for price in prices:
        total = 0
        total = total + price
    return total

print(total_price([10, 20, 30]))
```

No message at all. Where is the bug?

## Print debugging: make the intern talk

Add a `print` inside the loop to see what `total` is in every round.

```python
def total_price(prices):
    for price in prices:
        total = 0
        total = total + price
        print("total is", total)
    return total
print(total_price([10, 20, 30]))
```

**Found it!** `total` goes back to 0 every round.
**Fix:** move `total = 0` above the loop (and remove the debug `print`).

```python
def total_price(prices):
    total = 0
    for price in prices:
        total = total + price
    return total

print(total_price([10, 20, 30]))
```

## Crash or catch? `try` / `except`

Remember the Bug Zoo from Lecture 3? `int(input(...))` crashes when the user types `3.5` or `abc`.
One typo, and the **whole program dies**. A real app says *"please try again"* instead.

`try` / `except` lets you **catch** an error:
- `try:` "Intern, **try** this."
- An error inside `try`? Python jumps to `except`. The rest of `try` is **skipped**.
- No error? The `except` part is skipped.
- Same colon and indent as `if`.

Run the cell and type `20`. Run it again and type `abc`.

```python
text = input("Your age? ")
try:
    age = int(text)
    print(f"Next year: {age + 1}")
except ValueError:
    print("Please type a whole number")
```

Here we pretend the user typed `abc`, so you can see the `except` part without typing:

```python
text = "abc"   # pretend the user typed this
try:
    age = int(text)
    print(f"Next year: {age + 1}")
except ValueError:
    print("Please type a whole number")
```

## Ask until it's valid

`while True` repeats forever. `break` (Lecture 6) leaves the loop, and it runs **only** when `int(...)` worked.
Bad input? We jump to `except`, `break` is skipped, and the loop asks again.

Try typing `abc` first, then `20`.

```python
while True:
    try:
        age = int(input("Age? "))
        break
    except ValueError:
        print("Try again!")
print(f"Thanks! Age: {age}")
```

## 🐛 Bug Zoo: the missing key

```python
menu = {"falafel": 18, "tea": 8}
print(menu["Falafel"])
```

### What went wrong
Keys must match **exactly**: `"Falafel"` and `"falafel"` are different keys.

**Fix:** `menu["falafel"]`, or check first with `in`.

```python
menu = {"falafel": 18, "tea": 8}
print(menu["falafel"])
if "Falafel" in menu:
    print(menu["Falafel"])
else:
    print("No such dish")
```

## 🐛 Bug Zoo: looping over a dictionary

```python
menu = {"tea": 8, "coffee": 12}
for item, price in menu:
    print(item, price)
```

### What went wrong
Looping over `menu` alone gives only the **keys** (`"tea"`, `"coffee"`). Python then tries to split
the word `"tea"` into two boxes, `item` and `price`, and fails.

**Fix:** loop over `menu.items()` to get each key **with** its value.

```python
menu = {"tea": 8, "coffee": 12}
for item, price in menu.items():
    print(item, price)
```

## 🐛 Bug Zoo: the `except` that catches everything

Type `100` (a perfectly fine number). What does it print?

```python
try:
    price = int(input("Price? "))
    print(f"With VAT: {prise * 1.18}")
except:
    print("Please type a number")
```

### What went wrong
A bare `except:` (with no error name) catches **every** error, even our own typo `prise`.
The message blames the user, but the bug is ours!

**Fix:** always name the error you expect. Now Python shows the real problem, a `NameError`:

```python
try:
    price = int(input("Price? "))
    print(f"With VAT: {prise * 1.18}")
except ValueError:
    print("Please type a number")
```

Fix the typo too, and it works:

```python
try:
    price = int(input("Price? "))
    print(f"With VAT: {price * 1.18}")
except ValueError:
    print("Please type a number")
```
