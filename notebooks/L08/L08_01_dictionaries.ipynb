## Create, look up, change

Curly braces `{ }` build it. Square brackets with a key look it up.
A new key **adds** an item. An old key **updates** it (one price per dish!).

```python
menu = {"falafel": 18, "sabich": 24, "coffee": 12}
print(menu["sabich"])
menu["shakshuka"] = 38    # new key: add
menu["coffee"] = 14       # old key: update
print(menu)
```

## Is it there? `in` and `get`

`in` checks the **keys**. `get` gives a backup answer instead of crashing.

```python
phones = {"Noa": "050-1234567",
          "Omer": "052-7654321"}
print("Noa" in phones)
print("Yossi" in phones)
print(phones.get("Noa"))
print(phones.get("Yossi", "not saved"))
```

## Looping over a dictionary

`.items()` gives each key **with** its value. `.keys()` gives only keys, `.values()` only values.

```python
menu = {"falafel": 18,
        "sabich": 24,
        "coffee": 12}
for item, price in menu.items():
    print(f"{item}: {price} ILS")
print(sum(menu.values()))
```

## Counting votes: a tally board

The class votes for lunch. 🍕 Seen this food before? Add 1. New food? Start at 1.

```python
votes = ["pizza", "sushi", "pizza", "falafel", "pizza", "sushi"]
counts = {}
for food in votes:
    if food in counts:
        counts[food] = counts[food] + 1    # seen it: add 1
    else:
        counts[food] = 1                   # new: start at 1
print(counts)
```

## A shorter counter with `get`

`counts.get(r, 0)`: the count so far, or 0 if it's new. Same tally, one line instead of four. 👏

```python
reactions = ["laugh", "like", "laugh", "clap", "laugh", "like"]
counts = {}
for r in reactions:
    counts[r] = counts.get(r, 0) + 1
print(counts)
```
