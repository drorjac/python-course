## A table is a list of rows

`seats[1][2]`: **row** first, then **seat**. Counting starts at 0, as always.

```python
seats = [["Dana", "Omer", "Lior"],
         ["Noa", "Yael", "Tom"]]
print(seats[1])
print(seats[1][2])
print(len(seats))
```

## A loop inside a loop

Outer loop: one **student** (a row). Inner loop: each **grade** in that row.

```python
grades = [[90, 85, 70],    # Dana
          [60, 75, 81],    # Omer
          [100, 95, 88]]   # Lior
high = 0
for row in grades:
    for grade in row:
        if grade >= 85:
            high = high + 1
print("Grades of 85+:", high)
```

## A list of dictionaries

Like a spreadsheet: every row has **named** columns.

```python
students = [{"name": "Dana", "grade": 92},
            {"name": "Omer", "grade": 58},
            {"name": "Lior", "grade": 77}]
for student in students:
    if student["grade"] >= 60:
        print(student["name"], "passed")
```

## A matrix: a table with a clear shape

A **matrix** is a table where every row has the **same length**. Its shape is **rows x columns**.
`m[row][col]`: row first, then column. `[[1, 2, 3], [4, 5]]` is *not* a matrix: the rows have different lengths.

```python
m = [[1, 2, 3],
     [4, 5, 6]]
rows = len(m)
cols = len(m[0])
print(rows, "x", cols)
print(m[1][2])
```

## From a CSV file to a list of lists

A CSV file is a table in plain text: one line is one row, commas split the columns.
`split(",")` cuts each line into a row. Everything comes in as **text**, so use `int()` before doing math. 📒

```python
lines = ["Dana,90,85",
         "Omer,60,75",
         "Lior,100,95"]
table = []
for line in lines:
    table.append(line.split(","))
print(table[1])
print(table[2][0], int(table[2][1]) + 5)
```

## 🚀 Going further: matrix math with two loops: add

Same spot, add: `a[r][c] + b[r][c]`. Both matrices must have the **same shape**.
To **scale** by 3, use `3 * a[r][c]` instead.

```python
a = [[1, 2], [3, 4]]
b = [[10, 20], [30, 40]]
total = []
for r in range(len(a)):
    row = []
    for c in range(len(a[0])):
        row.append(a[r][c] + b[r][c])
    total.append(row)
print(total)
```

## 🚀 Going further: transpose: rows become columns

A 2 x 3 matrix becomes 3 x 2.

Words you will meet: the **diagonal** is `m[0][0]`, `m[1][1]`, ...; the **trace** is the sum of the diagonal;
the **identity** matrix has 1 on the diagonal and 0 everywhere else.

```python
m = [[1, 2, 3],
     [4, 5, 6]]
t = []
for c in range(len(m[0])):
    new_row = []
    for r in range(len(m)):
        new_row.append(m[r][c])
    t.append(new_row)
print(t)
```

## The average can lie

Five Tel Aviv apartments. One is a penthouse. 🏠 Let's compute the average rent.

```python
rents = [5800, 4500, 30000, 5200, 5000]
print(sum(rents) / len(rents))
```

Average: 10,100 ILS. Nobody pays that! The **median**, the middle value, tells a fairer story.

## The median by hand: sort, take the middle

5 items: `5 // 2` is 2, so we take position 2, the middle one.

```python
rents = [5800, 4500, 30000, 5200, 5000]
sorted_rents = sorted(rents)
print(sorted_rents)
middle = len(sorted_rents) // 2
print(sorted_rents[middle])
```

## A median function (odd *and* even)

With an even count there are two middle values, so we take their average.

```python
def median(values):
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:     # odd count
        return ordered[middle]
    else:                         # even count
        return (ordered[middle - 1] + ordered[middle]) / 2
```

```python
print(median([5800, 4500, 30000, 5200, 5000]))
print(median([3, 1, 4, 2]))
```

## Quantiles: where do you stand?

Sort the grades. The **0.25 quantile**: a quarter of the grades are at or below it.
The 0.5 quantile is just the median.

```python
def quantile(values, q):
    ordered = sorted(values)
    position = int(q * (len(ordered) - 1))
    return ordered[position]

grades = [88, 55, 74, 97, 62, 81, 70, 93, 78]
print(quantile(grades, 0.25))
print(quantile(grades, 0.5))
print(quantile(grades, 0.75))
```

One simple recipe (it rounds down). NumPy, in Lecture 9, has smoother ones.
