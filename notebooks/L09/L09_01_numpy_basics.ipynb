## 1. import: open the toolbox
A **library** is a box of ready-made tools someone else already wrote. `import` opens the box. `as np` gives it a short nickname, so we type `np` instead of `numpy`.

```python
import numpy as np

print(np.sqrt(16))
print(np.pi)
```

## 2. Meet the array
An **array** is a list built for math. It prints without commas, but indexes work exactly like lists (start at 0!).

```python
import numpy as np

prices = np.array([12, 15, 9, 20])
print(prices)
print(len(prices))
print(prices[0])
```

## 3. List vs array: same look, different powers
🔮 **Predict first:** what does `* 2` do to a list? And to an array?

```python
import numpy as np

prices_list = [12, 15, 9, 20]
prices = np.array([12, 15, 9, 20])
print(prices_list * 2)
print(prices * 2)
```

List `* 2` **repeats** the list. Array `* 2` **doubles every number**.

## 4. Add 17% VAT to every price
Your café receipt forgot the VAT. First the L06 way, with a loop:

```python
prices_list = [12, 15, 9, 20]
with_vat = []
for price in prices_list:
    with_vat.append(price * 1.17)
print(with_vat)
```

Now the NumPy way: one line. **Math on the whole array = math on every item.**

(The loop above shows tiny "computer rounding" crumbs like `17.549999999999997` instead of 17.55. Totally normal. NumPy prints them more neatly.)

```python
import numpy as np

prices = np.array([12, 15, 9, 20])
with_vat = prices * 1.17
print(with_vat)
```

## 5. Two arrays, item by item
Four students. Final grade = 30% homework + 70% exam. Item 1 goes with item 1, item 2 with item 2, and so on.

```python
import numpy as np

homework = np.array([90, 80, 100, 70])
exam = np.array([70, 85, 60, 95])

final = 0.3 * homework + 0.7 * exam
print(final)
```

## 6. Ask the array a question
`prices > 12` asks **every** price: True or False? Then `prices[...]` keeps only the `True` ones.

```python
import numpy as np

prices = np.array([12, 15, 9, 20])
print(prices > 12)
print(prices[prices > 12])
```

## 7. Statistics in one word each
Remember computing the median **by hand** in L08? Now it is one line. `percentile(grades, 90)`: 90% of the grades are below this number.

```python
import numpy as np

grades = np.array([55, 70, 88, 92, 100])
print(np.mean(grades))
print(np.median(grades))
print(np.percentile(grades, 90))
print(np.max(grades))
```

## 8. A 2-D array is a table
Rows are students (Dana, Omer, Noa), columns are tests (Test 1 to Test 4). `shape` tells you (rows, columns). `grades[row, col]` picks one cell, counting from 0 as always.

```python
import numpy as np

grades = np.array([[90, 80, 70, 60],
                   [60, 75, 85, 100],
                   [90, 100, 85, 89]])
print(grades.shape)
print(grades[1, 2])
print(grades.max())
```

## 9. Average per student, average per test
`axis=1` goes **across a row** (one student). `axis=0` goes **down a column** (one test).

```python
import numpy as np

grades = np.array([[90, 80, 70, 60],
                   [60, 75, 85, 100],
                   [90, 100, 85, 89]])
print(grades.mean(axis=1))   # each student (row)
print(grades.mean(axis=0))   # each test (column)
```

Dana 75, Omer 80, Noa 91. Test 2 was the easiest (85).
