## Bug 2: a list can't do VAT

```python
prices = [12, 15, 9, 20]
with_vat = prices * 1.17
```

**What went wrong:** `TypeError: can't multiply sequence by non-int of type 'float'`. A list can be *repeated* (`* 2`), but it can't do math on its items. And "repeat 1.17 times" makes no sense.

**Fix:** turn it into an array with `np.array(prices)`.

```python
prices = np.array([12, 15, 9, 20])
with_vat = prices * 1.17
print(with_vat)
```
