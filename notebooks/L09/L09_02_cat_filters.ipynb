# L09-02: Cat filters - a photo is just numbers

Lecture 9, slides: Load a real photo, Show the array as a picture, Filter 1-3, Photo filters are just math, Try it!, Predict the output

Zoom into any photo and you find tiny squares called **pixels**. Each pixel is 3 numbers: **R**ed, **G**reen, **B**lue, from 0 (none) to 255 (full). Meet Chelsea the cat 🐱: change her numbers and you change the photo. Every phone filter works like this.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## 1. Load a real photo: Chelsea the cat
`skimage` (scikit-image) is a library of photo tools that comes with sample images. `cat.shape` tells us (rows, columns, colors).

```python
from skimage import data

cat = data.chelsea()
print(cat.shape)
print(cat[150, 170])
```

300 rows of pixels, 451 columns, 3 colors per pixel. The pixel at row 150, column 170: lots of red, less green, little blue. That's the cat's fur!

## 2. Show the array as a picture
`plt.imshow` means "show this array as an image". (Matplotlib gets a full tour next week.)

```python
from skimage import data
import matplotlib.pyplot as plt

cat = data.chelsea()
plt.imshow(cat)
plt.show()
```

## 3. Filter 1: invert the colors
🔮 **Predict first:** `255 - cat` changes **every** number. What will a white whisker become?

```python
from skimage import data
import matplotlib.pyplot as plt

cat = data.chelsea()
inverted = 255 - cat
plt.imshow(inverted)
plt.show()
```

## 4. Filter 2: crop with slicing
Slicing from L05, now in 2-D: rows 70 to 210, columns 110 to 350.

```python
from skimage import data
import matplotlib.pyplot as plt

cat = data.chelsea()
eyes = cat[70:210, 110:350]
print(eyes.shape)
plt.imshow(eyes)
plt.show()
```

## 5. Filter 3: black and white
Average R, G and B for every pixel. `axis=2` is the color direction. One number per pixel = gray.

```python
from skimage import data
import matplotlib.pyplot as plt

cat = data.chelsea()
gray = cat.mean(axis=2)
print(gray.shape)
plt.imshow(gray, cmap="gray")
plt.show()
```

## 6. Photo filters are just math
Four more filters, one line each. A step of `-1` reverses, just like `"stressed"[::-1]` in L05.

**Upside down:** reverse the rows.

```python
plt.imshow(cat[::-1])
plt.show()
```

**Mirror:** keep all rows (`:`), reverse the columns.

```python
plt.imshow(cat[:, ::-1])
plt.show()
```

**Darker:** cut every number in half (`//` keeps whole numbers, like in L02).

```python
plt.imshow(cat // 2)
plt.show()
```

**Pure black & white:** ask every gray pixel "are you brighter than 100?" True = white, False = black.

```python
plt.imshow(gray > 100, cmap="gray")
plt.show()
```
