## 1. Every value is an object

Remember `type()` from Lecture 2? Every value has a type. Its official name is a **class**.
One value of that type is an **object**: `5` and `7` are two objects of class `int`.

```python
print(type(5))
print(type("hi"))
print(type([1, 2]))
print(type({"tea": 8}))
```

## 2. You have used methods all along

Data that belongs to an object is an **attribute**. An action that belongs to an object is a **method**:
`thing.action()`. Each class comes with its own methods: a list can `append`, a string can `upper`.

```python
cart = ["milk", "eggs"]
cart.append("pita")
print(cart)

name = "dana"
print(name.upper())

menu = {"tea": 8}
print(menu.get("cake", 0))
```

## 3. Why build your own?

Two points on a map, with what we know so far: four loose variables (easy to mix up),
or two dictionaries (`home["x"]` is long, and the actions live somewhere else).

```python
x1 = 0
y1 = 0
x2 = 3
y2 = 4
print(((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5)

home = {"x": 0, "y": 0}
work = {"x": 3, "y": 4}
print(((work["x"] - home["x"]) ** 2 + (work["y"] - home["y"]) ** 2) ** 0.5)
```

A **class** puts data *and* actions in one box. Think of a class as a **cookie cutter** 🍪:
each object is one cookie, same shape, its own toppings.

## 4. `class Point`: your own type

- `class Point:` the cookie cutter. Class names start with a capital letter.
- `Point(3, 4)` makes a new object, and Python runs `__init__` ("init" = start) for it.
- `self` means "this object". `self.x = x` stores `x` **inside** the object.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(3, 4)
print(p.x, p.y)
print(type(p))
```

## 5. Add a method: a function inside the class

A method is a `def` (Lecture 7) inside the class. Its first parameter is always `self`.
In `home.distance_to(work)`, `self` is `home` and `other` is `work`.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx ** 2 + dy ** 2) ** 0.5

home = Point(0, 0)
work = Point(3, 4)
print(home.distance_to(work))
```

## 6. Biology: a dish of bacteria 🔬

A method can **change** the object's data, and it can use loops and `if`s.
A toy model: the cells double every hour.

```python
class Culture:
    def __init__(self, name, cells):
        self.name = name
        self.cells = cells

    def grow(self, hours):
        for hour in range(hours):
            self.cells = self.cells * 2

    def report(self):
        print(f"{self.name}: {self.cells} cells")

dish = Culture("E. coli", 100)
dish.report()
dish.grow(3)
dish.report()
```

## 7. Money: a bank account in ILS 💰

Not every attribute has to be a parameter: `balance` always starts at 0.
The rules (no negative balance) live **inside** the class.

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money!")
        else:
            self.balance = self.balance - amount

account = BankAccount("Noa")
account.deposit(500)
account.withdraw(800)
account.withdraw(200)
print(account.owner, account.balance, "ILS")
```

## 🐛 Bug Zoo: the forgotten `self`

Inside `__init__`, `x = x` only makes a temporary local variable (Lecture 7: local scope).
Nothing is stored in the object.

```python
class Point:
    def __init__(self, x, y):
        x = x
        y = y

p = Point(3, 4)
print(p.x)
```

**Fix:** `self.x = x` stores it *in* the object. Here is the working `Point` again (run it, the cells below use it):

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx ** 2 + dy ** 2) ** 0.5

p = Point(3, 4)
print(p.x)
```

Second bug: a method without `self`.

```python
class Culture:
    def __init__(self, name, cells):
        self.name = name
        self.cells = cells

    def report():
        print(self.cells)

Culture("E. coli", 100).report()
```

**What went wrong:** Python always passes the object itself as the first argument. `report()` has no
room for it. **Fix:** `def report(self):`

```python
class Culture:
    def __init__(self, name, cells):
        self.name = name
        self.cells = cells

    def grow(self, hours):
        for hour in range(hours):
            self.cells = self.cells * 2

    def report(self):
        print(f"{self.name}: {self.cells} cells")

Culture("E. coli", 100).report()
```

```python
# Solution 2
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money!")
        else:
            self.balance = self.balance - amount

    def report(self):
        print(f"{self.owner} has {self.balance} ILS")

account = BankAccount("Noa")
account.deposit(300)
account.report()
```

```python
# Solution 3
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        return sum(self.grades) / len(self.grades)

dana = Student("Dana")
dana.add_grade(90)
dana.add_grade(80)
dana.add_grade(85)
print(dana.name, dana.average())
```

```python
# Solution 4
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx ** 2 + dy ** 2) ** 0.5

    def move(self, dx, dy):
        self.x = self.x + dx
        self.y = self.y + dy

p = Point(1, 1)
p.move(2, 3)
print(p.x, p.y)
```
