## 1. A DataFrame from a dictionary
Dictionary keys (L08) become **column names**. The numbers on the left are row numbers, called the **index**.

````python
import pandas as pd

menu = {"drink": ["Latte", "Tea", "Mocha"],
        "price": [16, 8, 18]}
df = pd.DataFrame(menu)
print(df)
````

Sun,Cappuccino,S,13,5
Sun,Espresso,S,9,5

## 3. Read the CSV, peek at the top
`head()` shows the first 5 rows. Always peek before you ask questions!

````python
import pandas as pd
df = pd.read_csv("coffee_orders.csv")
print(df.head())
````

## 4. One column = one array of numbers
`df["price"]` picks one column, like a dictionary key. Then the same statistics as NumPy.

````python
import pandas as pd

df = pd.read_csv("coffee_orders.csv")
print(df["price"].sum())
print(df["price"].mean())
print(df["price"].median())
print(df["price"].quantile(0.9))
````

Total spent: 538 ILS. Average order: 13.45 ILS.

## 5. Filter rows: only the lattes
Same trick as `prices[prices > 12]`: a True/False question for every row.

````python
import pandas as pd

df = pd.read_csv("coffee_orders.csv")
lattes = df[df["drink"] == "Latte"]
print(len(lattes))
print(round(lattes["price"].mean(), 2))
````

## 6. Filter rows: the expensive orders
🔮 **Predict first:** which drink will show up?

````python
import pandas as pd

df = pd.read_csv("coffee_orders.csv")
expensive = df[df["price"] >= 19]
print(expensive)
````

All four: large iced coffee. Notice that rows keep their original numbers (24, 30, ...).

## 7. Which drink is the most popular?
`value_counts()` counts each value, biggest first. Like the word counter from L08, in one word.

````python
import pandas as pd

df = pd.read_csv("coffee_orders.csv")
print(df["drink"].value_counts())
````

## 8. describe(): the whole summary at once
For every number column: count, mean, `std` (how spread out the values are), min, max and the quantiles `25%`, `50%`, `75%`.

````python
import pandas as pd

df = pd.read_csv("coffee_orders.csv")
print(df.describe())
````

## 9. 🔮 Predict the output
Back to the tiny menu. Which drinks are under 17, and what is their average price?

````python
import pandas as pd
menu = {"drink": ["Latte", "Tea", "Mocha"],
        "price": [16, 8, 18]}
df = pd.DataFrame(menu)
cheap = df[df["price"] < 17]
print(len(cheap))
print(cheap["price"].mean())
````
