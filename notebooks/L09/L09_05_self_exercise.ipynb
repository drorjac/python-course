### Solution 1: Menu prices as an array


```python
import numpy as np

prices = np.array([12, 10, 18, 25])
print(prices.mean(), prices.max(), prices.min())
discounted = prices * 0.9
print(discounted)
```

### Solution 2: A tiny class roster


```python
import pandas as pd

data = {'name': ['Noa', 'Tom', 'Omer', 'Lior'], 'score': [88, 41, 72, 56]}
df = pd.DataFrame(data)
print(df)
print('mean score:', df['score'].mean())
passing = df[df['score'] >= 56]
print(passing)
```
