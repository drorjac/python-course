### Solution 3: A thermometer class

````python
class Thermometer:
    def __init__(self):
        self.readings = []

    def add(self, degrees):
        self.readings.append(degrees)

    def highest(self):
        return max(self.readings)

    def has_fever(self):
        return self.highest() >= 38

patient = Thermometer()
patient.add(36.6)
patient.add(37.2)
patient.add(38.1)
print(patient.highest())
print(patient.has_fever())
````
