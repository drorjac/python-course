## 1. Six ways to compare
| Python | Means | Example (True) |
|---|---|---|
| `==` | is equal to | `5 == 5` |
| `!=` | is not equal to | `5 != 3` |
| `<` `>` | less / greater than | `3 < 5` |
| `<=` `>=` | "or equal" versions | `7 <= 7` |

Every comparison is a question. The answer: `True` or `False`.

```python
print(5 == 5)
print(5 != 3)
print(3 < 5)
print(5 > 3)
print(7 <= 7)
print(7 >= 7)
```

All six say `True`. Can you change **one** number so that a line says `False`?

## 2. `=` stores, `==` asks
- `age = 20`: "Put 20 in the box called `age`." 📦
- `age == 20`: "Is `age` equal to 20?" ❓

```python
age = 20             # one equals sign: store
print(age == 20)     # two equals signs: ask
print(age == 21)
```

## 4. Chaining: "between" in one line
Youth discount 🎟️: ages 18 to 30 get a cheaper ticket.
Read it like math class: "age is between 18 and 30".

```python
age = 23
print(18 <= age <= 30)
age = 45
print(18 <= age <= 30)
```

## 5. `and`: both must be True
| 18+? | On the list? | Gets in? |
|:---:|:---:|:---:|
| ✅ | ✅ | ✅ |
| ✅ | ❌ | ❌ |
| ❌ | ✅ | ❌ |
| ❌ | ❌ | ❌ |

A **truth table** lists every possible case and its answer.

```python
age = 20
on_list = True
if age >= 18 and on_list:
    print("Welcome to the party!")
else:
    print("Sorry, not tonight")
```

🔮 Change `age` to 17. Now one side is `False`, so the whole thing is `False`.

## 6. `or`: at least one must be True
| Birthday? | VIP? | Free entry? |
|:---:|:---:|:---:|
| ✅ | ✅ | ✅ |
| ✅ | ❌ | ✅ |
| ❌ | ✅ | ✅ |
| ❌ | ❌ | ❌ |

Only "no and no" stays out.

```python
birthday = False
vip = True
if birthday or vip:
    print("Free entry!")
else:
    print("Ticket: 50 ILS")
```

**Everyday trap:** in Python, `or` means "one, the other, **or both**".
("Tea or coffee?" in real life usually means just one.)

## 7. `not`: flip it
| `raining` | `not raining` |
|:---:|:---:|
| `True` | `False` |
| `False` | `True` |

```python
raining = False
print(not raining)
if not raining:
    print("Walk to campus")
```
