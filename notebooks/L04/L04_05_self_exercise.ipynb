### Solution 1: Leap year or not?

Check the rarest rule (400) first -- swap the order and 2000 comes out wrong.

```python
year = int(input("Year? "))
if year % 400 == 0:
    print(f"{year} is a leap year")
elif year % 100 == 0:
    print(f"{year} is not a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```
