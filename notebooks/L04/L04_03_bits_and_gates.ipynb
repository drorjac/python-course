A computer is billions of tiny switches. 💡 Each one is **on** (1, `True`) or **off** (0, `False`).
One switch is a **bit**, and 8 bits make a **byte**.
Let's peek at those 0s and 1s with Python.

**How to use:** run each cell with **Shift+Enter**. Before running, guess the output!

## 1. Counting with only 0 and 1
| Normal | Binary |
|:---:|:---:|
| 0 | `0` |
| 1 | `1` |
| 2 | `10` |
| 3 | `11` |
| 4 | `100` |
| 5 | `101` |

We count with 10 digits and write `10` after 9. Binary has only 0 and 1, so it writes `10` after 1.

## 2. Reading a binary number
Each place is worth **double** the one to its right: 8, 4, 2, 1.
Add up the places where the bulb is on. For `1101`: 8 + 4 + 1 = **13**.

```python
print(8 + 4 + 1)   # the bulbs that are on in 1101
```

## 3. Python speaks binary: `bin()`
- `0b` at the start just means "this is binary"
- `int("1101", 2)` goes back: binary text to a number

```python
print(bin(5))
print(bin(13))
print(bin(255))
print(int("1101", 2))
```

## 4. Logic gates: `and` / `or` as light switches
- **AND gate:** two switches in a row. Light only if A **and** B are closed.
- **OR gate:** two switches side by side. Light if A **or** B is closed.
- **NOT:** flips on to off, and off to on.

Chips are built from millions of these gates. Here they are in Python
(`True` = switch closed, `False` = switch open):

```python
switch_a = True
switch_b = False
print("AND light:", switch_a and switch_b)
print("OR light:", switch_a or switch_b)
print("NOT a:", not switch_a)
```

And the promise from the slides, 1 = `True` and 0 = `False`:

```python
print(int(True))
print(int(False))
```
