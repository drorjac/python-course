## 2. More than two roads: `elif`
Your exam grade: 90 or more is *Excellent*, 75 or more is *Very good*, 56 or more is *Passed*, otherwise *try again*. 🎓
- `elif` = "else, if..."
- add as many as you need
- one `else` at the end (optional)

```python
grade = 84
if grade >= 90:
    print("Excellent!")
elif grade >= 75:
    print("Very good")
elif grade >= 56:
    print("Passed")
else:
    print("Let's try again")
```

## 3. How the intern reads `elif`
1. `84 >= 90`? **False** → skip, check the next one
2. `84 >= 75`? **True** → print "Very good"
3. `84 >= 56`? never checked. `else`? never checked.

**Top to bottom. The first True wins. Then stop.**

## 6. Order matters!
🔮 **Predict first:** a student got 95. What does the program say?

```python
grade = 95
if grade >= 56:
    print("Passed")
elif grade >= 90:
    print("Excellent!")
```

A 95 that says only "Passed"? 😟 95 is also >= 56, so the first check wins and Python stops there.
No red error, the program just gives the wrong answer. Sneaky!

**Fix:** put the strictest check first.

```python
grade = 95
if grade >= 90:
    print("Excellent!")
elif grade >= 56:
    print("Passed")
```

## 7. Decisions inside decisions
An `if` can live inside another `if`. More spaces = one level deeper.
Saturday and raining? 📺 Saturday and sunny? 🏖️ Any other day? Class!

```python
day = "Saturday"
raining = True
if day == "Saturday":
    if raining:
        print("Netflix and blanket")
    else:
        print("Beach day!")
else:
    print("Go to class")
```
