## Bug 1: the forgotten colon
🔮 **Predict first:** what is missing?

```python
age = 20
if age >= 18
    print("Welcome")
```

**What went wrong:** every `if` line ends with a colon. Python says so: `SyntaxError: expected ':'`.

**Fix:** `if age >= 18:`

```python
age = 20
if age >= 18:
    print("Welcome")
```

## Bug 2: the forgotten indent
🔮 **Predict first:** the colon is there. What else is wrong?

```python
age = 20
if age >= 18:
print("Welcome")
```

**What went wrong:** the line under the `if` must be pushed right to show it **belongs to** the `if`:
`IndentationError: expected an indented block after 'if' statement on line 2`.

**Fix:** push `print` 4 spaces right.

```python
age = 20
if age >= 18:
    print("Welcome")
```

## Bug 3: `=` instead of `==`
🔮 **Predict first:** store or ask?

```python
age = 20
if age = 18:
    print("Exactly 18!")
```

**What went wrong:** inside an `if` you **ask** a question, so you need `==`. One `=` stores.
`SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?` (Ignore the `:=` part.)

**Fix:** `if age == 18:`. Now it runs. Nothing prints, because 20 is not 18. Change `age` to 18 and run again!

```python
age = 20
if age == 18:
    print("Exactly 18!")
```

## Bug 4: comparing text with a number
🔮 **Predict first:** run it and type `20`.

```python
age = input("Your age? ")
if age >= 18:
    print("Welcome")
```

**What went wrong:** `input()` gives text. `"20"` is text, `18` is a number. The intern is confused:
`TypeError: '>=' not supported between instances of 'str' and 'int'`.

**Fix:** convert right away: `age = int(input("Your age? "))`

```python
age = int(input("Your age? "))
if age >= 18:
    print("Welcome")
```
