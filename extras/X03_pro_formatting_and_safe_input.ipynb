## 1. More f-string tricks
After the `:` you can say **how** to show the value:

| Write | Means | Example result |
|---|---|---|
| `{x:,}` | commas in big numbers | `1,234,567` |
| `{x:.1%}` | as a percent, 1 digit | `17.5%` |
| `{x:05}` | pad with zeros to 5 digits | `00042` |
| `{x:>8}` | push right, 8 wide | `      14` |
| `{x:<12}` | push left, 12 wide | `Coffee      ` |
| `{x:^20}` | center, 20 wide | |

```python
visitors = 1234567
vat = 0.18
order = 42
print(f"Website visitors this year: {visitors:,}")
print(f"VAT: {vat:.1%}")
print(f"Order number {order:05}")
```

## 2. A receipt with straight columns

```python
title = "PIP'S CAFE"
print(f"{title:^28}")
print("-" * 28)
print(f"{'Coffee':<12}{2:>4}{28.0:>12.2f}")
print(f"{'Croissant':<12}{1:>4}{12.5:>12.2f}")
print(f"{'Cheesecake':<12}{1:>4}{24.9:>12.2f}")
print("-" * 28)
print(f"{'Total':<16}{65.4:>12.2f}")
```

Every bank statement, bus timetable and supermarket receipt is printed like this.
`<12` gives the name 12 spaces, `>4` and `>12.2f` push the numbers to the right edge so the decimal points line up.

## 3. Input you can trust
In Lecture 3, `int(input(...))` crashed when the user typed `twenty`. Real programs **check first**.
Text has built-in tools (called **methods**, more in Lecture 5). Three useful ones:

| Method | What it does | Example |
|---|---|---|
| `.strip()` | removes spaces at the start and end | `"  yes "` → `"yes"` |
| `.lower()` | small letters only | `"YES"` → `"yes"` |
| `.isdigit()` | `True` if it is only digits | `"20"` → `True`, `"twenty"` → `False` |

So the notebook runs on its own, the answer is written in a variable. Swap in `input(...)` to try it live.

```python
answer = "twenty"          # try: answer = input("Your age? ")
if answer.isdigit():
    age = int(answer)
    print(f"Next year you will be {age + 1}")
else:
    print(f"'{answer}' is not a number. Please use digits, like 20.")
```

```python
answer = "  YES "           # try: answer = input("Want a receipt? ")
clean = answer.strip().lower()
print(f"[{answer}] became [{clean}]")
if clean == "yes":
    print("Printing your receipt...")
else:
    print("No receipt. Have a nice day!")
```

**Real world:** "never trust user input" is rule number one of every web developer. A form that does not
check its input can crash, or worse: attackers type code instead of a name (look up "SQL injection").
Every sign-up form you ever filled checks your answers exactly like this.
