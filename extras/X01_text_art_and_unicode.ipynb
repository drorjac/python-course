## 1. Special characters: `\n` and `\t`
Inside quotes, a backslash `\` starts a **special character**:

| Write | Means |
|---|---|
| `\n` | new line |
| `\t` | tab (jump to the next column) |
| `\"` | a quote that does not end the text |
| `\\` | one real backslash |

🔮 **Predict first:** how many lines will this print?

```python
print("Pip's Cafe\nOpen 8:00-18:00\nWelcome!")
```

```python
print("Item\tPrice")
print("Coffee\t14")
print("Croissant\t12")
print("He said \"hi\"")
```

**Real world:** every file of data (CSV), every log file of a server, and every message your phone sends
is text with `\n` between the lines. The tab `\t` is how spreadsheets copy-paste tables.

## 2. Text on many lines: triple quotes
Three quotes `"""` let one piece of text go over many lines, exactly as you type it.
Programmers used this to draw pictures with letters, called **ASCII art**.

```python
print("""
   ( (
    ) )
  ........
  |      |]   Pip's Cafe
  \\      /    since 2026
   `----'
""")
```

## 3. Every language on Earth: Unicode
Python text is **Unicode**: one standard list with about 150,000 characters, from Hebrew and Arabic to
Chinese and emoji. Each character has a number. You can type it, or ask for it by its number with `\u`.

```python
print("שלום, Python!")
print("مرحبا  |  你好  |  Привет")
print("☕ 🥐 🐍")
print("\u2615")   # character number 2615 (in hexadecimal) is the coffee cup
```

**Real world:** before Unicode, a Hebrew email opened on another computer often showed up as gibberish
(`ùìåí` instead of `שלום`). Unicode (1991) fixed that, and today almost every website uses it.
Emoji are just Unicode characters too, that is why 🐍 looks the same on every phone.
