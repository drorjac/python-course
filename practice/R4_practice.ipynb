### Solution 9: Predict the taxi fare

10 + 4 x 12 = 58 and 10 + 4 x 30 = 130. The data sits exactly on a line, so the model finds it exactly.
30 km is outside the examples, but taxi fares really do keep growing in a line, so here it works.

````python
from sklearn.linear_model import LinearRegression

km = [[2], [5], [8], [10], [15], [20]]
fare = [18, 30, 42, 50, 70, 90]
model = LinearRegression()
model.fit(km, fare)
print(model.predict([[12], [30]]))
````

### Solution 10: Spam or not?

[3, 4] sits in the middle: its 3 nearest examples are all "ok". Close calls are where models make mistakes.

````python
from sklearn.neighbors import KNeighborsClassifier

emails = [[0, 0], [1, 1], [0, 2], [2, 1],
          [7, 9], [5, 12], [8, 6], [6, 10]]
labels = ["ok", "ok", "ok", "ok",
          "spam", "spam", "spam", "spam"]
model = KNeighborsClassifier(n_neighbors=3)
model.fit(emails, labels)
print(model.predict([[6, 8], [1, 0]]))
print(model.predict([[3, 4]]))
````
