### Solution 1: fix the tip calculator

A colon `:` after the `def` line; `return`, not `print` (otherwise `print(tip(200))` shows `None`); a comma between the two values.

````python
def tip(bill, percent=10):
    return bill * percent / 100

print(tip(200))
print(tip(150, 15))
````

### Solution 2: bigger, and even?

If `a > b` we leave with `a`. Otherwise we reach the last line and return `b`: no `else` needed. `n % 2 == 0` is already `True` or `False`.

````python
def bigger(a, b):
    if a > b:
        return a
    return b

def is_even(n):
    return n % 2 == 0

print(bigger(7, 12))
print(is_even(7), is_even(12))
````

### Solution 3: weather in Fahrenheit

The function does one job. We call it once per day and keep the answer in `f`.

````python
def c_to_f(celsius):
    return celsius * 9 / 5 + 32

week = [18, 24, 31, 15]
for c in week:
    f = c_to_f(c)
    if f > 80:
        print(f"{c}C = {f}F  hot!")
    else:
        print(f"{c}C = {f}F")
````

### Solution 8: fix the median

`//`, not `/`: a position must be a whole number. Sort first. Even count: average the two middle values.

````python
def median(values):
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2

print(median([30, 10, 20]))
print(median([30, 10, 20, 40]))
````

### Solution 9: average per student

The loop inside the loop is now hidden in the function. Each row is just a list.

````python
def row_average(row):
    total = 0
    for value in row:
        total = total + value
    return total / len(row)

grades = [[90, 85, 70],
          [60, 75, 81],
          [100, 95, 88]]
names = ["Dana", "Omer", "Lior"]
for i in range(len(grades)):
    print(f"{names[i]}: {row_average(grades[i]):.1f}")
````

### Solution 10: library late fees

One loop, three jobs: the fee, the running total, and the biggest so far.

````python
def late_fee(days):
    if days <= 14:
        return 0
    return (days - 14) * 2

loans = [{"name": "Noa", "days": 10},
         {"name": "Itay", "days": 21},
         {"name": "Maya", "days": 30}]
total = 0
worst = loans[0]
for loan in loans:
    fee = late_fee(loan["days"])
    total = total + fee
    if loan["days"] > worst["days"]:
        worst = loan
    print(f"{loan['name']}: {fee} ILS")
print(f"Total fees: {total} ILS")
print(f"Owes the most: {worst['name']}")
````

### Solution 11: the safe divider

`float("abc")` raises `ValueError`, `12 / 0.0` raises `ZeroDivisionError`. Each `except` **returns** a message,
so the caller always gets an answer. Never a bare `except:`: it would hide our own typos too.

````python
def safe_divide(a_text, b_text):
    try:
        return float(a_text) / float(b_text)
    except ValueError:
        return "Not a number!"
    except ZeroDivisionError:
        return "Can't divide by 0!"

print(safe_divide("12", "4"))
print(safe_divide("12", "abc"))
print(safe_divide("12", "0"))
````
