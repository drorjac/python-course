### Solution 8: Shekels to dollars

The `if` replaces the answer only for small amounts. Try 500 and 50.

````python
shekels = float(input("How many shekels? "))
rate = 3.7
dollars = shekels / rate
if shekels < 100:
    dollars = (shekels - 5) / rate
print(f"You get {dollars:.2f} dollars")
````
