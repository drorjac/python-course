### Solution 1: Fix the parking sign

`==` to ask, not `=`; a colon `:` after the condition; indent the `else` line by 4 spaces. Friday is not Saturday, but 19 > 18 is True: `or` needs just one.

````python
hour = 19
day = "Friday"
if day == "Saturday" or hour > 18:
    print("Parking is free")
else:
    print("Pay at the machine")
````

### Solution 2: Low battery warning

`and not charging`, then `elif` and `else`. With `charging = True`, `not charging` is False, so the first check fails. 12 < 50 is True: `Okay for now`.

````python
battery = 12
charging = False
if battery < 15 and not charging:
    print("Plug me in NOW!")
elif battery < 50:
    print("Okay for now")
else:
    print("All good")
````

### Solution 3: Leap year or not?

Rarest rule first. Swap the order and 2000 comes out wrong!

````python
year = int(input("Year? "))
if year % 400 == 0:
    print(f"{year} is a leap year")
elif year % 100 == 0:
    print(f"{year} is not a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
````

### Solution 4: Shabbat dinner shopping

`append` adds to the end, `remove` crosses off by name, `insert` cuts in at a position, `len` counts.

````python
cart = ["challah", "wine", "hummus"]
cart.append("salad")
cart.remove("wine")    # we have some
cart.insert(0, "candles")
print(cart, len(cart))
````

### Solution 5: Read the train ticket

`HFA` sits at 4, 5, 6, so the stop is 7. `[8:]` also works for the time.

````python
ticket = "TLV-HFA-0830"
origin = ticket[:3]
dest = ticket[4:7]
departs = ticket[-4:]
print(f"From {origin} to {dest}")
print(f"Leaves at {departs[:2]}:{departs[2:]}")
````

### Solution 6: What should Yoni watch?

Sorting first means the output is the same on every computer. Sets alone have no order.

````python
noa = {"Fauda", "Friends", "Shtisel", "Lupin"}
yoni = {"Friends", "Dark", "Fauda", "Narcos"}
print(f"They share {len(noa & yoni)} shows")
for_yoni = list(noa - yoni)
for_yoni.sort()
print(f"For Yoni: {for_yoni}")
print(f"In total: {len(noa | yoni)} shows")
````

### Solution 7: The even-floor elevator

### Solution 8: Fix the tips average

`total = 0` goes **before** the loop, `print` goes **after** (no indent), and divide by `len(tips)`: 5, not 4. The buggy version printed 5.0, 8.75, 0.0, 12.5, 3.75.

````python
tips = [20, 35, 0, 50, 15]
total = 0
for tip in tips:
    total = total + tip
print(f"Average tip: {total / len(tips)}")
````

### Solution 9: Saving for a festival

450 < 450 is False, so the loop stops at exactly 450.

````python
savings = 150
weeks = 0
while savings < 450:
    savings = savings + 75
    weeks = weeks + 1
    print(f"Week {weeks}: {savings} ILS")
print(f"Ticket after {weeks} weeks!")
````

### Solution 10: Password checker

Three lectures in one program! Try it with `telaviv2026` and `ab12`.

````python
password = input("Password? ")
digits = 0
for letter in password:
    if letter in "0123456789":
        digits = digits + 1
print(f"Digits: {digits}")
if len(password) >= 8 and digits >= 2:
    print("Strong password!")
else:
    print("Weak password")
````
